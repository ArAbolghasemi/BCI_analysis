# Learning the task, the intuitive axis, and a new one

The behavioural half of the story: the animal learns to drive a tone with its own population
activity, and — once the decoder is silently rotated mid-session — finds and exploits a new
population axis, then returns to the old one when it is restored.

### What the numbers turned out to say

Worth knowing before reading the figures, because two of these are not what the section titles
would lead you to expect. Every one is re-measured by the checks at the end.

1. **Acquisition takes one session, not weeks.** `mouse_0403L` is at a 90 % hit rate on its
   first BMI session; within that session the median time to target falls by about half (§2.1).
   `mouse_0322R` starts at 24 % and is in its normal band by the fourth session.
2. **After that there is no across-day learning curve at all** — no trend in hit rate or in
   time to target on either animal (§3.1, §3.2). What varies across days is motivation:
   performance tracks body weight significantly and session number not at all (§3.3), and it
   swings by more than that *within* a single hour (§3.4).
3. **The rotation costs the animal something only the first time it meets one.** On the first
   remapping session the hit rate falls 100 % → 74 % and the median time to target goes
   1.1 s → 6.2 s over the rotated block; from the second session on there is no detectable
   cost, and pooled over all fifteen sessions a paired test finds nothing (§5.2).
4. **The washout is immediate**, on that same first session, and the new axis is a genuinely
   different direction — 75–86° from the intuitive one, where two random directions in this
   population would be 88° apart (§5.3, §5.5). On that first session a simulation of the
   *inactive* axis reads above the target more often than the cursor the animal was actually
   driving — one of only two sessions in fifteen where that happens (§5.1b).

### What the figures are made of

**`CURSOR_REP` in the setup cell chooses how the cursor is drawn**, everywhere at once:

| value | what is plotted | axis |
|---|---|---|
| `"hz"` *(default)* | `freqToOutput`, the tone the speaker played | **log** |
| `"oct"` | the same signal as $\log_2(f/1\,\text{kHz})$ | linear |
| `"brain"` | `cursor_brain`, the decoder's own gated readout $D$ | linear |
| `"theta"` | $D/\theta$, so the target is exactly 1.0 everywhere | linear |

The Hz axis is **logarithmic on purpose**. The tone law is exponential, so on a linear Hz axis
the entire negative half of the cursor's range — everything between 1000 and 2963 Hz, which
is 20–45 % of all frames — is squashed into the bottom eighth of the panel and reads as a flat
line sitting on 2963 Hz. That is an axis artefact, not a floor in the data; `"brain"` and
`"theta"` show the same signal with zero in the middle and no saturation at all.

The reward tone is drawn with everything else, and the law is anchored to each day's own
threshold, which is what makes one axis legitimate across animals and days: the measured
cursor → tone law

$$\log_2 f \;=\; \log_2(2962.635\ \text{Hz}) \;+\; 1.5669 \cdot \frac{D}{\theta}$$

($R^2 = 1.000000$ on 66/66 single-threshold mouse-days) is anchored to **each day's own reward
threshold** $\theta$. A cursor at $D = \theta$ is therefore **8777.2 Hz on every animal and
every day** — verified below — even though $\theta$ itself is 1.5 or 1.9 on `mouse_0403L`, 1.5
throughout on `mouse_0322R`, and 0.9 for the rotated axis. In `"brain"` units that same target
is a **step** at the block boundary, so the reward reference is drawn as a per-frame line
rather than an `axhline`. The tone saturates at 1000 Hz and 15107.6 Hz.

The law is used in the other direction exactly once: in §5.1, to turn a *simulated* cursor into
the tone it would have produced.

### Colour

`intuitive axis = #7f3f98`, `new axis = #f7941d`, everywhere, set once in the setup cell.
Time within a trial is viridis and session number is a separate ramp (`cool`) that stays clear
of both axis colours. Anything that does not encode a variable is grey.

Outcome is **hit** green / **miss** red `#e8453c` — deliberately *not* the Okabe-Ito
vermillion the rest of this repo uses, because vermillion and the new-axis orange are the same
hue and the two families meet in §5.5 and in the movies. Red against green is the usual
colourblind trap, so wherever both appear the outcome is **also** coded by marker and dash.
Nothing here says "correct" or "incorrect": a trial is a **hit** or a **miss**.

**Markers mean the same thing in every trajectory panel: a square is where a trial started and
a circle is where it ended.** Where hits and misses are drawn together, a hit ends on a circle
and a miss on a cross. $T_{int}$ / $T_{new}$ are crosses-of-the-X-kind and the quiescence
baseline is a diamond, so neither can be read as a trial endpoint.

### Assumptions and caveats

* **The two animals are not equivalent.** `mouse_0403L` is the primary animal: 36 sessions, one
  decoder throughout, 15 remapping days. `mouse_0322R` **has no remapping days at all** (one
  `factor_to_use` per session) and its decoder was **redefined on 2023-05-02** — the first six
  BMI sessions used a different one. Day 0 for it is therefore taken as `20230502`, which is
  the convention in `first_pass_analysis.ipynb`, and the earlier sessions are excluded. Only
  the across-day summaries (§3) use both animals.
* **A miss is censored, not measured.** `duration_trial` is **20 s** on every day of both
  animals, and 99.5 % of failed trials last exactly 19.97 s. Trial duration is therefore
  reported for **hits only**, as a time-to-target; the miss distribution carries nothing beyond
  the hit rate.
* **Smoothing is display-only** and uses the *same* kernel as the neural trajectories
  (`SMOOTH_SEC`, a 0.33 s Gaussian = 2 bins at 6 Hz). It is applied with a normalised
  convolution so the out-of-session gaps neither spread nor get filled.
* **The first session of `mouse_0403L` used $\theta = 1.5$**, every later session 1.9. On the
  tone axis that is invisible (the target is 8777 Hz either way), but it does mean the first
  day was *easier in cursor units*, so §2.1's within-session comparison is sound while a
  day-1-against-day-2 comparison is not.
* **Time on task confounds every within-session comparison.** The reward rate falls over an
  hour (§3.4), and in a remapping session the washout block always sits 50–75 min in. A block
  that happens late is not a matched control for one that happens early.
* **`factor_to_use` labels the axis, not the number 1.** On `mouse_0403L` the intuitive axis is
  **factor 2** and the rotated one **factor 3**; factor 1 appears only in `mouse_0322R`'s
  pre-redefinition sessions. Nothing here hard-codes a factor number — the intuitive axis is
  "whatever the session started on", which is how the rig's own block sequence defines it.
* **The rotation time is the end of the last intuitive-axis trial** (`na.remapping_start_frame`),
  not the first new-axis trial start: the new block begins the moment the old one stops. It
  lands at 10.6–11.1 min on every remapping day.
* **Day 0 has no BMI trials** on either animal (it is the decoder-definition session), so
  "excluding day 0" costs nothing — it is excluded by the trial filter anyway.
* **The population axis in §4 is estimated, not read off the packet.** The packet's stored
  weights are badly truncated (378/468 neurons carry one; ~62 % of the factor's squared norm
  survives), so the decoder direction is re-estimated each day by regressing the rig's own
  recovered $M$ on that day's binned activity. §4.2 checks it against the stored weights.
* **A PC1–PC2 plane is a view, not a summary.** This population needs ~260 components to reach
  80 % of its variance, and the decoder axis puts less than a third of itself in that plane.
  Decoder-axis arrows are therefore drawn as **directions at a fixed length**, with the
  in-plane fraction reported separately.
* Three sessions of `mouse_0403L` (20230728, 20230730, 20230731) are 80 min long but their
  **loggers stop at 60 min**. Alignment holds from frame 0; the traces simply end early.

### 1.1 · Loading

One cell holds every loader. `day_table` is the per-session summary that §3 plots; `session`
returns the per-frame tone trace that §2 and §5 draw, cached so a figure cell re-runs instantly.
`swarm_offsets` is the deterministic bee-swarm used wherever single trials are shown.

The tone constants are **measured**, not configured: `HZ_REWARD` is the frequency the reward
override clamps to, and the check cell at the end verifies it is the same 8777.2 Hz on every
animal and day, which is what makes one shared y-axis legitimate.

```python
# ---- the cursor, in whichever representation CURSOR_REP selects ----
# The rig's chain is  M = X.F  ->  CS_k = M_k/||M||  ->  D_k = |CS_k|^p M_k  ->  tone, and the
# logger stores both ends of it. So there are four honest ways to draw "the cursor", and they
# are the same signal seen through more or less of that chain:
#
#   "hz"    freqToOutput, the tone the speaker played, in Hz. Drawn on a LOG axis, because the
#           tone law is exponential: on a linear Hz axis everything below 2963 Hz is squashed
#           into the bottom eighth of the panel and the whole negative half of the cursor's
#           range looks like a flat line sitting on 2963.
#   "oct"   the same signal as log2(f/1 kHz) -- a linear perceptual axis.
#   "brain" cursor_brain, the decoder's own gated readout D. No tone saturation, no reward
#           clamp, and zero is zero. This is the one to use to judge the simulation.
#   "theta" D divided by that block's reward threshold, so the target is exactly 1.0 on every
#           animal, day and block.
#
# hz/oct come from `freqToOutput`, brain/theta from `cursor_brain`; the two differ on ~7 % of
# frames (the rig's own clipping) and only `freqToOutput` carries the reward clamp.
TONE_F0, TONE_SLOPE = 2962.635, 1.5669        # log2 f = log2 F0 + SLOPE * D/theta
HZ_BASE   = TONE_F0                           # 2963 Hz: the tone at cursor 0
HZ_REWARD = 8777.2                            # the tone at cursor = theta, i.e. the reward clamp
HZ_FLOOR, HZ_CEIL = 1000.0, 15107.6           # where the tone saturates, on every animal
d_to_hz = lambda d, thr: np.clip(TONE_F0 * 2.0 ** (TONE_SLOPE * np.asarray(d, float) / thr),
                                 HZ_FLOOR, HZ_CEIL)

REPS = {
    "hz":    dict(label="cursor (Hz)", scale="log", lim=(950.0, 16500.0),
                  ticks=[1000, 2963, 8777, 15108],
                  ticklabels=["1000", "2963", "8777", "15108"]),
    "oct":   dict(label="cursor (octaves re 1 kHz)", scale="linear", lim=(-0.10, 4.00),
                  ticks=[0.0, 1.5669, 3.1338, 3.9173],
                  ticklabels=["0", "1.57", "3.13", "3.92"]),
    "brain": dict(label="cursor ($D$)", scale="linear", lim=(-2.6, 4.2),
                  ticks=None, ticklabels=None),
    "theta": dict(label=r"cursor ($D/\theta$)", scale="linear", lim=(-1.4, 2.2),
                  ticks=[-1, 0, 1, 2], ticklabels=["-1", "0", "1", "2"]),
}
if CURSOR_REP not in REPS:
    raise ValueError(f"CURSOR_REP must be one of {sorted(REPS)}, got {CURSOR_REP!r}")
REP = dict(REPS[CURSOR_REP], name=CURSOR_REP)


def d_to_rep(d, thr):
    """A decoder value D (with its own block's threshold) in the active representation."""
    if REP["name"] == "brain":
        return np.asarray(d, float)
    if REP["name"] == "theta":
        return np.asarray(d, float) / np.asarray(thr, float)
    hz = d_to_hz(d, thr)
    return hz if REP["name"] == "hz" else np.log2(hz / 1000.0)


def rep_axis(ax, ylim=None, ticks=True, label=None):
    """Scale, limits, ticks and y label for an axis drawn in the active representation."""
    ax.set_yscale(REP["scale"])
    ax.set_ylim(*(REP["lim"] if ylim is None else ylim))
    if ticks and REP["ticks"] is not None:
        ax.set_yticks(REP["ticks"])
        ax.set_yticklabels(REP["ticklabels"])
        ax.minorticks_off()
    if label is not False:
        ax.set_ylabel(REP["label"] if label is None else label, fontsize=12 * FONT_SCALE)
    return ax


def rep_frac(f, ylim=None):
    """A height given as a fraction of the cursor axis, honouring a log scale."""
    lo, hi = (REP["lim"] if ylim is None else ylim)
    if REP["scale"] == "log":
        return float(np.exp(np.log(lo) + f * (np.log(hi) - np.log(lo))))
    return float(lo + f * (hi - lo))


def draw_reward_level(ax, S, t, f0, f1, **kw):
    """The reward reference across the drawn window.

    It is a per-frame line, not an axhline: on a remapping day each block carries its own
    reward threshold (1.9 intuitive, 0.9 rotated), so in `brain` units this is a step."""
    kw = dict(dict(color=P["correct"], lw=1.4, zorder=1), **kw)
    return ax.plot(t, S.rew_level[f0:f1], **kw)


R    = {m: na.open_mouse(rfs, DATA_ROOT, m) for m in MICE}
DAYS = {m: na.get_day_list(R[m]) for m in MICE}
EXP_LOG = next((p for p in [REPO / "RH + GH experiment log.xlsx",
                            DATA_ROOT / "RH + GH experiment log.xlsx"] if p.exists()), None)


def day_zero_index(mouse):
    """Index of the session that counts as day 0 for this animal."""
    d0 = DAY0_OVERRIDE.get(mouse) or R[mouse]["date_day0"].load()
    return DAYS[mouse].index(d0)


def smooth_nan(x, sigma):
    """Gaussian smooth that neither spreads a NaN nor fills one.

    Normalised convolution (signal and mask smoothed together), then the NaNs are put back
    exactly where the sample itself was missing -- so the out-of-session gap at the end of a
    truncated logger stays one gap instead of eating several seconds of real trace."""
    x = np.asarray(x, float); ok = np.isfinite(x)
    if sigma <= 0:
        return np.where(ok, x, np.nan)
    num = gaussian_filter1d(np.where(ok, x, 0.0), sigma, mode="nearest")
    den = gaussian_filter1d(ok.astype(float), sigma, mode="nearest")
    out = np.divide(num, den, out=np.full_like(num, np.nan), where=den > 1e-6)
    out[~ok] = np.nan
    return out


_SESS = {}


def session(mouse, day_i):
    """The per-frame behaviour of one session: the cursor the animal drove, its trials, and
    the block structure. The cursor is NaN only outside the experiment."""
    key = (mouse, day_i, REP["name"])
    if key in _SESS:
        return _SESS[key]
    r, day = R[mouse], DAYS[mouse][day_i]
    dec, ts = ca.get_decoder_df(r, day), ca.get_timeseries_df(r, day)
    nf = min(len(dec), len(ts))                      # 3 sessions have a truncated logger
    f   = np.asarray(dec["freqToOutput"], float)[:nf]
    D   = np.asarray(dec["cursor_brain"], float)[:nf]
    ovr = np.asarray(ts["frequencyOverride"], float)[:nf] > 0.5
    inx = np.asarray(ts["CE_experimentRunning"], float)[:nf] > 0.5
    thr = np.asarray(ts["cursor_threshold_reward"], float)[:nf]
    hz_raw = np.where(inx & (f > 0), f, np.nan)      # Hz as played, reward clamp included
    if REP["name"] == "hz":
        raw, rew_level, base_level = hz_raw, np.full(nf, HZ_REWARD), HZ_BASE
    elif REP["name"] == "oct":
        raw = np.log2(hz_raw / 1000.0)
        rew_level, base_level = np.full(nf, np.log2(HZ_REWARD / 1000.0)), np.log2(HZ_BASE / 1000.0)
    elif REP["name"] == "brain":
        raw = np.where(inx, D, np.nan)
        rew_level, base_level = thr, 0.0             # a step at the block boundary
    else:                                            # "theta"
        raw = np.where(inx, D / thr, np.nan)
        rew_level, base_level = np.ones(nf), 0.0
    tr = na.get_true_bmi_trials(na.get_trials_df(r, day))
    S = SimpleNamespace(
        mouse=mouse, i=day_i, day=day, nf=nf, rep=REP["name"],
        cur=smooth_nan(raw, SMOOTH_SEC * SR), cur_raw=raw,
        rew_level=rew_level, base_level=base_level,
        hz=smooth_nan(hz_raw, SMOOTH_SEC * SR), hz_raw=hz_raw,
        ovr=ovr, in_exp=inx,
        quiet=np.asarray(ts["CS_quiescence"], float)[:nf] > 0.5,
        trials=tr, rew=na.get_active_reward_frames(r, day),
        remap=na.remapping_start_frame(tr), blocks=na.segment_factor_blocks(tr),
        thr=float(np.nanmedian(thr)), thr_all=np.unique(np.round(thr[np.isfinite(thr)], 3)),
        thr_by_f=ca.reward_threshold_by_factor(r, day))
    _SESS[key] = S
    return S


def day_table(mouse):
    """One row per session: trial counts, hit rate, time-to-target, block structure, weight."""
    r, days = R[mouse], DAYS[mouse]
    w = na.load_weight_percentage(EXP_LOG, mouse, days) if EXP_LOG else {d: np.nan for d in days}
    z = day_zero_index(mouse)
    rows = []
    for i, d in enumerate(days):
        tr = na.get_true_bmi_trials(na.get_trials_df(r, d))
        rec = dict(i=i, day=d, rel=i - z, n=len(tr), weight=100.0 * w.get(d, np.nan))
        if len(tr):
            hit = tr.success == 1
            dur = (tr.end - tr.start) / SR
            fac = tr.factor[np.isfinite(tr.factor)]
            ts  = ca.get_timeseries_df(r, d)
            thr = np.unique(np.round(np.asarray(ts["cursor_threshold_reward"], float), 3))
            rec.update(n_hit=int(hit.sum()), hit_rate=float(hit.mean()),
                       t_hit=float(np.nanmedian(dur[hit])) if hit.any() else np.nan,
                       t_hit_lo=float(np.nanpercentile(dur[hit], 25)) if hit.any() else np.nan,
                       t_hit_hi=float(np.nanpercentile(dur[hit], 75)) if hit.any() else np.nan,
                       factors=",".join(f"{v:g}" for v in np.unique(fac)),
                       is_remap=bool(np.unique(fac).size > 1),
                       remap_f=na.remapping_start_frame(tr),
                       thr=",".join(f"{v:g}" for v in thr[np.isfinite(thr)]),
                       n_rew=len(na.get_active_reward_frames(r, d)))
        rows.append(rec)
    t = pd.DataFrame(rows)
    t["is_remap"] = t["is_remap"].fillna(False).astype(bool)
    return t


TAB = {m: day_table(m) for m in MICE}
# the days each animal contributes to the learning curve: after day 0, has trials, pre-rotation
LEARN = {m: TAB[m][(TAB[m]["rel"] > 0) & (TAB[m]["n"] > 0) & ~TAB[m]["is_remap"]]
            .reset_index(drop=True) for m in MICE}
REMAP_I = {m: TAB[m].loc[TAB[m]["is_remap"], "i"].to_numpy() for m in MICE}
HAS_TRIALS = {m: TAB[m].loc[TAB[m]["n"] > 0, "i"].to_numpy() for m in MICE}


def rel_day(mouse, i):
    """Session number counted from that animal's day 0 -- the number the figures label."""
    return int(TAB[mouse].set_index("i").loc[int(i), "rel"])


def swarm_offsets(y, width, nbins=34):
    """Deterministic bee-swarm offsets: points sharing a y-bin are laid out side by side.

    Jitter hides density and overplots; a swarm shows it, and it is identical on every run."""
    y = np.asarray(y, float)
    off = np.zeros(len(y))
    g = np.isfinite(y)
    if g.sum() < 2:
        return off
    lo, hi = np.nanmin(y[g]), np.nanmax(y[g])
    edges = np.linspace(lo, hi + 1e-9, nbins + 1)
    b = np.clip(np.digitize(y, edges) - 1, 0, nbins - 1)
    counts = np.bincount(b[g], minlength=nbins)
    step = 2.0 * width / max(counts.max() - 1, 1)
    for k in np.flatnonzero(counts):
        j = np.flatnonzero(g & (b == k))
        j = j[np.argsort(y[j])]
        off[j] = (np.arange(len(j)) - (len(j) - 1) / 2.0) * step
    return off


def square_limits(axes, pad=0.06, force=None):
    """One square frame for a row of trajectory panels.

    Every panel gets the SAME span on x and on y -- so a centimetre is a centimetre in both
    directions and the shapes are comparable -- centred on the data rather than anchored to a
    shared corner, which would otherwise leave half of each panel empty."""
    if force is not None:
        return tuple(force), tuple(force)
    xs = [a.get_xlim() for a in axes]; ys = [a.get_ylim() for a in axes]
    x0, x1 = min(v[0] for v in xs), max(v[1] for v in xs)
    y0, y1 = min(v[0] for v in ys), max(v[1] for v in ys)
    h = (1 + pad) * max(x1 - x0, y1 - y0) / 2.0
    cx, cy = (x0 + x1) / 2.0, (y0 + y1) / 2.0
    return (cx - h, cx + h), (cy - h, cy + h)


print(f"cursor representation: {REP['name']} — {REP['label']}, {REP['scale']} axis")
for m in MICE:
    print(f"{m}: {len(DAYS[m])} sessions, day 0 = {DAYS[m][day_zero_index(m)]} "
          f"(index {day_zero_index(m)}), {len(LEARN[m])} learning days, "
          f"{len(REMAP_I[m])} remapping days, thresholds {sorted(set(LEARN[m]['thr']))}")
```

## 2 · What a session looks like

### 2.1 · The first session, early / middle / late

The cursor the animal drove, in whatever `CURSOR_REP` is set to — by default the tone in
Hz, on a **log** axis. The target sits at a fixed **8777 Hz** and the cursor's own zero at
**2963 Hz**, both constants of the task rather than of the day. Trials are shaded by outcome,
hits are marked where the reward fired and misses where the trial timed out. The flat tops at
8777 Hz are the **reward tone**: during `frequencyOverride` the speaker is clamped there for
about a second regardless of the population, so those samples are the reward, not cursor
control. `REWARD_TONE = "dim"` draws them in grey if you want them separated, and
`CURSOR_REP = "brain"` removes them entirely, because the decoder signal is never clamped.

This is where the learning is. On `mouse_0403L`'s first BMI session the **early** window shows
the animal parked at the baseline tone through trial after trial, every one of them running to
the timeout; by the **late** window the trace is a train of sharp excursions that reach the line
and come straight back down. The cell prints the numbers by thirds of the session — the median
time to target roughly halves — and §3 shows that by the next session the across-day curve is
already flat. Acquisition in this task is a matter of one session, not of weeks.

**Caveat for this day only:** `mouse_0403L`'s first session used $\theta = 1.5$ rather than the
1.9 of every later session. On the tone axis that is invisible — the law is anchored to
$\theta$ — but the animal needed *less* population drive on this day than on any later one, so
the within-session comparison above is sound while a day-1-against-day-2 comparison is not.

```python
# ===== SETTINGS =====
DAY_I       = 1             # index into the mouse's day list; 1 = the first session with trials
WIN_MIN     = 3.0           # length of each window, in minutes
STARTS_MIN  = None          # explicit [t0, t1, t2] in minutes, or None -> early / middle / late
STAGE_AT    = (0.04, 0.50, 0.96)      # where in the trial-bearing part of the session to look
STAGE_LAB   = ("early", "middle", "late")
YLIM        = None          # None = the active representation's own limits
REWARD_TONE = "draw"        # "draw" = one trace | "dim" = clamp frames in grey | "hide"
SHADE_ALPHA = 0.16
LW          = 1.1
FIGSIZE     = (13.5, 7.6)
TITLE       = None
FIGNAME     = "fig01_first_session_stages"
# ====================
S = session(MOUSE, DAY_I)
span = (float(np.nanmin(S.trials.start)), float(np.nanmax(S.trials.end)))
win  = WIN_MIN * 60 * SR
starts = ([m * 60 * SR for m in STARTS_MIN] if STARTS_MIN is not None else
          [np.clip(span[0] + a * (span[1] - span[0]) - win / 2, 0, S.nf - win) for a in STAGE_AT])
y_hit, y_miss = rep_frac(0.95, YLIM), rep_frac(0.03, YLIM)

fig, axes = plt.subplots(len(starts), 1, figsize=FIGSIZE, sharey=True, constrained_layout=True)
for ax, f0, lab in zip(np.atleast_1d(axes), starts, STAGE_LAB):
    f0 = int(f0); f1 = int(min(f0 + win, S.nf))
    t  = np.arange(f0, f1) / SR / 60.0
    dw.shade_trials(ax, S.trials.start, S.trials.end, S.trials.success,
                    colors={1.0: C_HIT, 0.0: C_MISS}, alpha=SHADE_ALPHA,
                    sr=SR * 60.0, t0=0.0, clip=(f0, f1))
    draw_reward_level(ax, S, t, f0, f1, lw=1.6)
    ax.axhline(S.base_level, color=P["muted"], lw=1.0, ls=":")
    cur = S.cur[f0:f1]
    if REWARD_TONE in ("dim", "hide"):           # split the reward clamp off the cursor trace
        o = S.ovr[f0:f1]
        if REWARD_TONE == "dim":
            ax.plot(t, np.where(o, cur, np.nan), color=P["reward_tone"], lw=LW, zorder=1.5)
        cur = np.where(o, np.nan, cur)
    ax.plot(t, cur, color=P["cursor"], lw=LW, zorder=2)
    dw.mark_events(ax, S.rew, y_hit, sr=SR * 60.0, t0=0.0, clip=(f0, f1),
                   marker="v", s=34, color=C_HIT, zorder=5)
    miss = S.trials.end[(S.trials.success == 0) & np.isfinite(S.trials.end)]
    dw.mark_events(ax, miss, y_miss, sr=SR * 60.0, t0=0.0, clip=(f0, f1),
                   marker="x", s=34, color=C_MISS, lw=1.6, zorder=5)
    ax.set_xlim(t[0], t[-1])
    rep_axis(ax, YLIM)
    ax.set_title(f"{lab}  ({t[0]:.0f}–{t[-1]:.0f} min)", loc="left",
                 fontsize=12.5 * FONT_SCALE, color=P["muted"], pad=4)
np.atleast_1d(axes)[-1].set_xlabel("time in the session (min)")
leg = [("hit", "patch", C_HIT, dict(alpha=SHADE_ALPHA)),
       ("miss", "patch", C_MISS, dict(alpha=SHADE_ALPHA)),
       ("reward delivered", "marker", C_HIT, dict(marker="v", ms=7)),
       ("timeout", "marker", C_MISS, dict(marker="x", ms=7)),
       ("reward threshold", "line", P["correct"], dict(lw=2.2)),
       ("cursor 0", "line", P["muted"], dict(lw=1.4, ls=":"))]
if REWARD_TONE == "dim":
    leg.insert(4, ("reward clamp", "line", P["reward_tone"], dict(lw=1.6)))
dw.legend_from(np.atleast_1d(axes)[0], leg, loc="lower center", bbox_to_anchor=(0.5, 1.14),
               ncol=len(leg), fontsize=11 * FONT_SCALE)
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()

hit = S.trials.success == 1
dur = (S.trials.end - S.trials.start) / SR
edges = span[0] + np.array([0, 1 / 3, 2 / 3, 1.0]) * (span[1] - span[0])
print(f"{S.day} (session {rel_day(MOUSE, DAY_I)}), by thirds of the session:")
for k in range(3):
    m = (S.trials.start >= edges[k]) & (S.trials.start < edges[k + 1])
    print(f"  third {k + 1}: n = {m.sum():3d} trials, hit rate {np.mean(hit[m]):.0%}, "
          f"median time to target {np.nanmedian(dur[m & hit]):.2f} s")
g = np.isfinite(S.hz_raw)
print(f"tone on this session: median {np.nanmedian(S.hz_raw):.0f} Hz, "
      f"{np.mean(S.hz_raw[g] >= HZ_REWARD):.1%} of in-session frames at or above the reward "
      f"tone, {np.mean(S.hz_raw[g] >= HZ_CEIL - 1):.1%} at the 15108 Hz ceiling, "
      f"{np.mean(S.hz_raw[g] < HZ_BASE):.1%} BELOW the tone at cursor 0 — that last "
      f"number is the half of the range a linear Hz axis hides")
```

## 3 · Learning across days

### 3.1 · Hit rate

One bar per session, day 0 excluded and remapping days excluded, so every bar is the animal
working a single fixed decoder for a whole hour.

**This is not a learning curve, and that is the result.** `mouse_0403L` hits **90 % on its very

So the acquisition happens **inside the first one to three sessions** (§2.1 shows it happening

### 3.2 · Time to target

The duration of a **hit** is the time from trial onset to the threshold crossing, so it measures
how directly the animal drove the population to the target — a graded measure where the hit
rate is at ceiling. Misses are not plotted: every one of them lasts 19.97 s, because
`duration_trial` is 20 s and a failed trial simply runs out. Their distribution is a spike at
the ceiling (dashed line) and carries nothing the hit rate does not already say.

One row per animal, because a readable swarm needs the full width. Every point is one trial,
placed by a **deterministic bee-swarm** rather than random jitter — the horizontal spread *is*
the local density, and it is identical on every run. Boxes are the quartiles over that
session's hits, whiskers the 10th–90th percentiles.

Read with §3.1: the median time to target is **also flat across sessions** on both animals, and
the spread within a session is far larger than any difference between sessions. The one place
it does move is *within* the first session (§2.1).

```python
# ===== SETTINGS =====
N_SWARM  = 120         # trials drawn per session (fixed seed); None = all of them
WIDTH    = 0.34        # half-width of the swarm, in session units
DOT      = dict(s=7, alpha=0.8, lw=0)
BOX_W    = 0.74
YLIM     = (0, 21)
FIGSIZE  = (14.0, 9.0)   # one row per animal: the swarm needs the full width to be readable
TITLE    = None
FIGNAME  = "fig04_time_to_target"
# ====================
rng = np.random.default_rng(0)
fig, axes = plt.subplots(len(MICE), 1, figsize=FIGSIZE, constrained_layout=True)
for ax, m, letter in zip(np.atleast_1d(axes), MICE, "abcd"):
    t = LEARN[m]
    data, pos = [], []
    for _, row in t.iterrows():
        tr = session(m, int(row["i"])).trials
        d = ((tr.end - tr.start) / SR)[tr.success == 1]
        d = d[np.isfinite(d)]
        data.append(d); pos.append(int(row["rel"]))
        k = d if (N_SWARM is None or len(d) <= N_SWARM) else rng.choice(d, N_SWARM, replace=False)
        ax.scatter(row["rel"] + swarm_offsets(k, WIDTH), k, zorder=4,
                   color=P["fg"], **DOT)
    bp = ax.boxplot(data, positions=pos, widths=BOX_W, showfliers=False, whis=(10, 90),
                    patch_artist=True, zorder=3,
                    medianprops=dict(color=P["task"], lw=2.8),
                    boxprops=dict(facecolor=P["task"], edgecolor=P["task"], alpha=0.30, lw=0),
                    whiskerprops=dict(color=P["task"], lw=1.6),
                    capprops=dict(color=P["task"], lw=1.6))
    ax.axhline(20.0, color=P["incorrect"], lw=1.4, ls="--")
    ax.annotate("trial timeout (a miss)", xy=(pos[-1], 20.0), xytext=(0, -16),
                textcoords="offset points", ha="right", va="top",
                color=P["incorrect"], fontsize=11 * FONT_SCALE)
    ax.text(0.5, 1.01, m.replace("mouse_", ""), transform=ax.transAxes, ha="center",
            va="bottom", fontsize=13.5 * FONT_SCALE, color=P["fg"])
    ax.set_xlabel("session (days after day 0)")
    ax.set_ylabel("time to target (s)")
    ax.set_xticks(pos); ax.set_xticklabels([str(p) for p in pos])
    ax.set_xlim(pos[0] - 0.8, pos[-1] + 0.8)
    ax.set_ylim(*YLIM)
    dw.label_panel(ax, letter, color=P["fg"])
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()
for m in MICE:
    t = LEARN[m]
    sl = stats.linregress(t["rel"], t["t_hit"])
    n = sum(int(np.sum(session(m, int(i)).trials.success == 1)) for i in t["i"])
    print(f"{m}: {n} hits over {len(t)} sessions; median time to target "
          f"{t['t_hit'].head(3).mean():.1f} s over the first 3 days vs "
          f"{t['t_hit'].tail(3).mean():.1f} s over the last 3; "
          f"slope {sl.slope:+.3f} s/day (p = {sl.pvalue:.1g})")
```

### 3.3 · Body weight and performance

If skill is not what varies across days, motivation is the obvious candidate, and body weight is
the standard proxy for it — a percentage of free-feeding weight, taken from
`RH + GH experiment log.xlsx`. One panel letter per **column**, since a column is one analysis
and the two rows are the two animals: **a** overlays weight on the hit rate day by day, **b**
and **c** are the scatters with an ordinary least-squares fit.

On `mouse_0403L` the correlation is real and in the expected direction — a heavier (less
hungry) animal is slower and misses more — and it **strengthens** when session number is
partialled out, which is the opposite of what a spurious drift-over-time correlation would do.
`mouse_0322R` points the same way more weakly.

Two honest limits. The weight range is **narrow** (about 2.5 percentage points on `0403L`), so
these fits are extrapolating very little, and a per-session body weight is a crude stand-in for
how motivated the animal was in any particular minute — §3.4 shows the same animal's motivation
swinging within a single hour.

```python
# ===== SETTINGS =====
FIGSIZE  = (14.0, 7.4)
TITLE    = None
FIGNAME  = "fig05_weight_vs_performance"
# ====================
# One panel letter per COLUMN: a column is one analysis, and the two rows are the two animals.
fig, axes = plt.subplots(len(MICE), 3, figsize=FIGSIZE, constrained_layout=True)
axes = np.atleast_2d(axes)
for row, m in zip(axes, MICE):
    t = LEARN[m].dropna(subset=["weight"])
    # (a) the overlay: weight and hit rate against session number
    ax = row[0]
    ax.plot(t["rel"], t["weight"], color=P["baseline"], marker="o", ms=4, lw=1.8)
    ax.set_ylabel("body weight (% free-feeding)", color=P["baseline"], fontsize=12 * FONT_SCALE)
    ax.tick_params(axis="y", colors=P["baseline"])
    tw = ax.twinx(); tw.spines["right"].set_visible(True)
    tw.plot(t["rel"], t["hit_rate"], color=P["task"], marker="s", ms=4, lw=1.8)
    tw.set_ylabel("hit rate", color=P["task"], fontsize=12 * FONT_SCALE)
    tw.tick_params(axis="y", colors=P["task"]); tw.set_ylim(0, 1.05)
    ax.set_xlabel("session (days after day 0)")
    ax.text(-0.26, 0.5, m.replace("mouse_", ""), transform=ax.transAxes, ha="center",
            va="center", rotation=90, fontsize=13.5 * FONT_SCALE, color=P["fg"])
    # (b, c) the scatters
    for ax, col, lab in [(row[1], "hit_rate", "hit rate"),
                         (row[2], "t_hit", "median time to target (s)")]:
        x, y = t["weight"].to_numpy(), t[col].to_numpy()
        g = np.isfinite(x) & np.isfinite(y)
        ax.scatter(x[g], y[g], s=34, color=P["task"], alpha=0.8, lw=0)
        sl = stats.linregress(x[g], y[g])
        xs = np.array([x[g].min(), x[g].max()])
        ax.plot(xs, sl.slope * xs + sl.intercept, color=P["null"], lw=2.0, ls="--")
        # put the statistics in the corner the fit line leaves empty
        yx, va = ((0.04, "bottom") if sl.slope < 0 else (0.96, "top"))
        ax.annotate(f"r = {sl.rvalue:+.2f}\np = {sl.pvalue:.2g}", xy=(0.03, yx),
                    xycoords="axes fraction", va=va, ha="left",
                    fontsize=11 * FONT_SCALE, color=P["muted"])
        ax.set_xlabel("body weight (% free-feeding)")
        ax.set_ylabel(lab, fontsize=12 * FONT_SCALE)
for ax, letter in zip(axes[0], "abc"):
    dw.label_panel(ax, letter, color=P["fg"])
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()


def partial_r(x, y, z):
    """Correlation of x and y with z regressed out of both."""
    g = np.isfinite(x) & np.isfinite(y) & np.isfinite(z)
    rx = x[g] - np.polyval(np.polyfit(z[g], x[g], 1), z[g])
    ry = y[g] - np.polyval(np.polyfit(z[g], y[g], 1), z[g])
    return stats.pearsonr(rx, ry)


for m in MICE:
    t = LEARN[m].dropna(subset=["weight"])
    print(f"{m}: weight spans {t['weight'].min():.1f}–{t['weight'].max():.1f} % "
          f"over {len(t)} sessions")
    for col, lab in [("hit_rate", "hit rate"), ("t_hit", "time to target")]:
        r0 = stats.pearsonr(t["weight"], t[col])
        rp = partial_r(t["weight"].to_numpy(), t[col].to_numpy(), t["rel"].to_numpy().astype(float))
        print(f"   vs {lab}: r = {r0[0]:+.2f} (p = {r0[1]:.2g}); "
              f"with session number partialled out r = {rp[0]:+.2f} (p = {rp[1]:.2g})")
```

### 3.4 · Reward rate within a session

Motivation is not constant inside an hour either. Active rewards counted in a sliding
`WIN_MIN`-minute window, stepped every `STEP_MIN`; the highlighted session is `DAY_I`, the grey
traces are every other learning day of the same animal, so it is clear whether the shape is
typical or that one day.

The usual shape is a decline with pauses — the animal works, stops, and works again. That is
the thing to keep in mind for every session-average in this notebook, and it is the reason §5's
later blocks have to be read with a grain of salt: a block that happens at minute 60 of an
80-minute session is not comparable to one at minute 5, whatever decoder it is running.

```python
# ===== SETTINGS =====
DAY_I    = 6            # the highlighted session
WIN_MIN  = 2.5          # sliding window length
STEP_MIN = 0.5          # step between windows
SHOW_ALL = True         # all other learning days, behind, in grey
FIGSIZE  = (12.0, 4.4)
TITLE    = None
FIGNAME  = "fig06_reward_rate_within_session"
# ====================
fig, ax = plt.subplots(figsize=FIGSIZE, constrained_layout=True)


def reward_rate(m, i):
    S = session(m, i)
    w = na.sliding_windows(S.nf, win_sec=WIN_MIN * 60, step_sec=STEP_MIN * 60, sr=int(SR))
    return (np.array([x.center_sec for x in w]) / 60.0,
            na.reward_rate_in_windows(S.rew, w, sr=int(SR)))


if SHOW_ALL:
    for i in LEARN[MOUSE]["i"]:
        if int(i) == DAY_I:
            continue
        x, y = reward_rate(MOUSE, int(i))
        ax.plot(x, y, color=P["single"], lw=1.0, alpha=0.7, zorder=1)
x, y = reward_rate(MOUSE, DAY_I)
ax.plot(x, y, color=P["task"], lw=2.6, zorder=3)
ax.axhline(np.mean(y), color=P["task"], lw=1.2, ls="--", zorder=2)
S = session(MOUSE, DAY_I)
ax.set_xlim(0, S.nf / SR / 60.0)
ax.set_xlabel("time in the session (min)")
ax.set_ylabel("active rewards per min")
dw.legend_from(ax, [(f"day {rel_day(MOUSE, DAY_I)}", "line", P["task"], dict(lw=2.6)),
                    ("every other learning day", "line", P["single"], dict(lw=1.2)),
                    ("its session mean", "line", P["task"], dict(lw=1.2, ls="--"))],
               loc="lower left", fontsize=11 * FONT_SCALE)
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()
# a peak/trough ratio is meaningless when the trough is a window with no reward at all,
# so report the swing as a fraction of the session spent well below its own mean instead
below = np.mean(y < 0.5 * np.mean(y))
print(f"day {rel_day(MOUSE, DAY_I)}: {np.mean(y):.1f} rewards/min on average, "
      f"peak {np.max(y):.1f}, trough {np.min(y):.1f}; {below:.0%} of the session is spent "
      f"below half that mean")
cv, lo = [], []
for i in LEARN[MOUSE]["i"]:
    _, yy = reward_rate(MOUSE, int(i))
    cv.append(np.std(yy) / np.mean(yy) if np.mean(yy) > 0 else np.nan)
    lo.append(np.mean(yy < 0.5 * np.mean(yy)))
print(f"across the {len(cv)} learning days: within-session coefficient of variation "
      f"{np.nanmedian(cv):.2f} (median), and a median {np.nanmedian(lo):.0%} of each session "
      f"below half its own mean rate")
```

## 4 · The population behind it

Activity is `spks` from the packet, binned to **6 Hz** (`BIN` frames) and z-scored per neuron
over the session — the same preprocessing every other notebook in this repo uses. The cell
below makes **two pooled passes** over it, each cached: one over the stable-control sessions,
which §4.1 and §4.2 draw in, and one over the remapping sessions, which §5 uses. Each pass
accumulates the covariance instead of holding the sessions, so memory stays at
(neurons × neurons) however many days go in, and it also returns, per session, the mean
trajectory over hits and the decoder axis estimated inside each block.

### 4.1 · One space, three sessions

Panel **a** is why the trajectory panels have to be read carefully: this population is **not**
low-dimensional. PC1 carries under 8 % of the variance and it takes hundreds of components to
reach 80 %. A PC1–PC2 plane is a legitimate *view*, not a summary — the printed participation
ratio says how many dimensions are really in play.

Panel **b** is individual hits from three sessions — the first, the tenth and the
twentieth — all in the **same** space, the one fitted on every stable-control day together, so
the three panels are directly comparable rather than three separate PCAs. Each trial runs from
onset to the threshold crossing and is coloured by time within the trial; the path is smoothed
for display with the same 0.33 s kernel as the cursor. The panels are square and share one range
on both axes. `TRAJ_DAYS`, `N_TRIALS`, `PCS` and `LIM` are where to experiment.

```python
# ===== SETTINGS =====
POOL_I    = None      # day indices for the stable space; None = every learning day of MOUSE
REMAP_II  = None      # day indices for the remapping space; None = every remapping session
N_WARP    = 40        # bins each hit is warped to before averaging
RECOMPUTE = False     # True forces the pooled passes to run again instead of using the cache
# ====================
# TWO passes over the neural data, each cached: one over the stable-control sessions (4.1,
# 4.2) and one over the remapping sessions (5.x). The covariance is accumulated across
# sessions instead of holding them, so memory stays (neurons x neurons) however many days
# are pooled. Everything downstream reads these arrays; nothing reloads the neural data.


def binned(mouse, i):
    """(neurons, bins) z-scored activity, and the trial table in bin units."""
    r, d = R[mouse], DAYS[mouse][i]
    X = na.bin_and_zscore(na.load_neural(r, d), BIN)
    return X, na.bin_trials(na.get_true_bmi_trials(na.get_trials_df(r, d)), BIN)


def factor_bin_mask(mouse, i, nb, factor):
    """Bins during which `factor` was the live decoder (all bins when factor is None)."""
    if factor is None:
        return np.ones(nb, bool)
    S = session(mouse, i)
    m = np.zeros(nb, bool)
    for b in S.blocks:
        if b.factor == factor:
            e = b.end_frame if np.isfinite(b.end_frame) else S.nf
            m[int(b.start_frame // BIN):int(min(e, S.nf) // BIN)] = True
    return m


def recovered_M(mouse, i, nb):
    """The rig's own gate-free readout M = D/|CS|^p, averaged into the neural bins.

    `decoder_angles` is CS and `cursor_brain_raw` is D, so this inverts the gate exactly;
    |CS| < 0.05 is masked because there M = D/CS^2 is 0/0."""
    dec = ca.get_decoder_df(R[mouse], DAYS[mouse][i])
    D = np.asarray(dec["cursor_brain_raw"], float)
    CS = np.asarray(dec["decoder_angles"], float)
    ok = np.isfinite(D) & np.isfinite(CS) & (np.abs(CS) > 0.05)
    Mf = np.where(ok, D / np.abs(CS) ** 2.0, np.nan)
    n = min(nb, len(Mf) // BIN)
    out = np.full(nb, np.nan)
    out[:n] = np.nanmean(Mf[:n * BIN].reshape(n, BIN), axis=1)
    return out


def pooled_pass(mouse, idx, tag, factors=(None,), n_warp=None, recompute=False):
    """Pooled covariance, plus -- per day and per requested factor -- the mean
    trajectory over hits and the decoder axis estimated inside that factor's own block.

    Returns mu, cov, traj (days, factors, n_warp, neurons), axis (days, factors, neurons),
    r2 (days, factors) and the day indices."""
    n_warp = N_WARP if n_warp is None else n_warp
    fp = CACHE / f"pooled_{tag}.npz"
    if fp.exists() and not recompute:
        z = np.load(fp)
        return z["mu"], z["cov"], z["traj"], z["axis"], z["r2"], z["idx"]
    cov = sm = None; nt = 0
    traj, axis, r2 = [], [], []
    for i in idx:
        X, tr = binned(mouse, int(i))
        if cov is None:
            cov = np.zeros((X.shape[0], X.shape[0])); sm = np.zeros(X.shape[0])
        cov += X @ X.T; sm += X.sum(axis=1); nt += X.shape[1]
        Mb = recovered_M(mouse, int(i), X.shape[1])
        per_f_t, per_f_a, per_f_r = [], [], []
        for f in factors:
            sel = (tr.success == 1) if f is None else ((tr.success == 1) & (tr.factor == f))
            segs = []
            for j in np.flatnonzero(sel):
                a, b = int(tr.start[j]), min(int(tr.end[j]), X.shape[1])
                if b - a >= 3:
                    segs.append(na.resample_trajectory(X[:, a:b].T, n_warp))
            per_f_t.append(np.mean(segs, axis=0) if segs
                           else np.full((n_warp, X.shape[0]), np.nan))
            # the decoder axis: least squares from the binned activity to the rig's own M
            g = np.isfinite(Mb) & factor_bin_mask(mouse, int(i), X.shape[1], f)
            if g.sum() < 50:
                per_f_a.append(np.full(X.shape[0], np.nan)); per_f_r.append(np.nan); continue
            A, y = X[:, g].T, Mb[g] - np.nanmean(Mb[g])
            beta = np.linalg.lstsq(A, y, rcond=None)[0]
            per_f_a.append(beta / np.linalg.norm(beta))
            per_f_r.append(1.0 - np.var(y - A @ beta) / np.var(y))
        traj.append(per_f_t); axis.append(per_f_a); r2.append(per_f_r)
    mu = sm / nt
    cov = cov / nt - np.outer(mu, mu); cov *= nt / (nt - 1)
    out = (mu, cov, np.asarray(traj), np.asarray(axis), np.asarray(r2), np.asarray(idx))
    np.savez_compressed(fp, mu=out[0], cov=out[1], traj=out[2], axis=out[3],
                        r2=out[4], idx=out[5])
    return out


def eig_space(cov):
    """Descending eigenvalues, eigenvectors and the variance ratio of a pooled covariance."""
    val, vec = np.linalg.eigh(cov)
    o = np.argsort(val)[::-1]
    val, vec = val[o], vec[:, o]
    pos = np.clip(val, 0, None)
    return val, vec, pos / pos.sum()


# ---- which factor is which on this animal, read from the block structure, never hard-coded
_S0 = session(MOUSE, int(REMAP_I[MOUSE][0]))
F_INT = float(_S0.blocks[0].factor)          # the axis every session starts on
F_NEW = float(_S0.blocks[1].factor)          # the axis the rig rotates to mid-session

# ---- pass 1: the stable-control space ----
POOL = [int(i) for i in (LEARN[MOUSE]["i"] if POOL_I is None else POOL_I)]
MU, COV, MTRAJ, AXIS, AXR2, POOL_IDX = pooled_pass(
    MOUSE, POOL, f"{MOUSE}_stable{len(POOL)}_b{BIN}_w{N_WARP}", (None,), N_WARP, RECOMPUTE)
MTRAJ, AXIS, AXR2 = MTRAJ[:, 0], AXIS[:, 0], AXR2[:, 0]
VAL, VEC, VRAT = eig_space(COV)
CUM = np.cumsum(VRAT)
PR_POOL = VAL.sum() ** 2 / (VAL ** 2).sum()
DIM80 = int(np.searchsorted(CUM, 0.80) + 1)

# ---- pass 2: the remapping space, one trajectory and one axis per block ----
RD = [int(i) for i in (REMAP_I[MOUSE] if REMAP_II is None else REMAP_II)]
MUr, COVr, TRAJr, AXr, AXR2r, RIDX = pooled_pass(
    MOUSE, RD, f"{MOUSE}_remap{len(RD)}_b{BIN}_w{N_WARP}", (F_INT, F_NEW), N_WARP, RECOMPUTE)
VALr, VECr, VRATr = eig_space(COVr)
CUMr = np.cumsum(VRATr)
AXI, AXN = AXr[:, 0], AXr[:, 1]              # intuitive-block and rotated-block estimates
ang = np.degrees(np.arccos(np.clip(np.abs(np.sum(AXI * AXN, axis=1)), 0, 1)))
chance = np.degrees(np.arccos(np.sqrt(2 / (np.pi * len(MUr)))))

# ---- how the estimate compares with the packet's own (truncated) weight column ----
W_PKT = np.asarray(R[MOUSE]["weights_day0_idxFactors"].load(), float)
HAVE = np.isfinite(W_PKT).all(axis=1)
w_store = np.where(HAVE, W_PKT[:, int(F_INT) - 1], 0.0); w_store /= np.linalg.norm(w_store)
cos_store = np.abs(AXIS[:, HAVE] @ w_store[HAVE]) / np.linalg.norm(w_store[HAVE])
cos_pc1 = np.abs(AXIS @ VEC[:, 0])
in_plane = np.linalg.norm(AXIS @ VEC[:, :2], axis=1)

print(f"the intuitive axis is factor {F_INT:g}, the rotated one factor {F_NEW:g}")
print(f"stable pool: {len(POOL_IDX)} sessions, {len(MU)} neurons | PC1 {VRAT[0]:.1%}, "
      f"PC1-2 {CUM[1]:.1%}, {DIM80} PCs to 80 %, participation ratio {PR_POOL:.0f}")
print(f"remap pool:  {len(RIDX)} sessions | PC1 {VRATr[0]:.1%}, PC1-4 {CUMr[3]:.1%}, "
      f"participation ratio {VALr.sum() ** 2 / (VALr ** 2).sum():.0f}")
print(f"decoder-axis fit to the rig's own M: stable R² {AXR2.min():.2f}–{AXR2.max():.2f}, "
      f"remapping blocks R² {np.nanmin(AXR2r):.2f}–{np.nanmax(AXR2r):.2f}")
print(f"|cos| with the packet's stored weights {cos_store.min():.2f}–{cos_store.max():.2f} "
      f"(truncation ceiling {np.sqrt(0.62):.2f}); with pooled PC1 "
      f"{cos_pc1.min():.2f}–{cos_pc1.max():.2f} (chance {1/np.sqrt(len(MU)):.3f}); "
      f"{in_plane.min():.0%}–{in_plane.max():.0%} of it lies in the PC1–PC2 plane")
print(f"angle between the intuitive and rotated estimates: {ang.min():.0f}–{ang.max():.0f} deg "
      f"(median {np.median(ang):.0f}) against {chance:.0f} deg for two random directions")
```

```python
# ===== SETTINGS =====
TRAJ_DAYS = [1, 10, 20]        # sessions to draw, as indices into the mouse's day list
N_TRIALS  = 12                 # hits drawn per panel (fixed seed)
PCS       = (0, 1)             # which two components of the POOLED space to draw
N_SPEC    = 30                 # components shown in the spectrum
LIM       = None               # (lo, hi) applied to BOTH axes of every PCA panel; None = auto
SEED      = 0
FIGSIZE   = (15.5, 4.6)
TITLE     = None
FIGNAME   = "fig07_single_day_pca"
# ====================
# Every trajectory panel lives in the SAME space -- the one fitted on all stable-control days
# together in the cell above -- so the panels are directly comparable, not three separate PCAs.
VK = VEC[:, list(PCS)]


def trial_paths(X, tr, n=None, seed=0):
    """Projected, display-smoothed paths for HIT trials, in the pooled space."""
    ok = np.flatnonzero((tr.success == 1) & np.isfinite(tr.start) & np.isfinite(tr.end))
    if n is not None and len(ok) > n:
        ok = np.sort(np.random.default_rng(seed).choice(ok, n, replace=False))
    out = []
    for j in ok:
        a, b = int(tr.start[j]), min(int(tr.end[j]), X.shape[1])
        if b - a < 3:
            continue
        p = (X[:, a:b].T - MU) @ VK
        out.append(gaussian_filter1d(p, SMOOTH_BIN, axis=0, mode="nearest"))
    return out


fig, axes = plt.subplots(1, 1 + len(TRAJ_DAYS), figsize=FIGSIZE, constrained_layout=True)
ax = axes[0]
k = np.arange(1, N_SPEC + 1)
ax.bar(k, VRAT[:N_SPEC] * 100, color=P["muted"], width=0.8)
tw = ax.twinx(); tw.spines["right"].set_visible(True)
tw.plot(k, CUM[:N_SPEC] * 100, color=P["fg"], lw=2.0)
tw.set_ylabel("cumulative (%)", fontsize=12 * FONT_SCALE); tw.set_ylim(0, 100)
ax.set_xlabel("component"); ax.set_ylabel("variance explained (%)")
ax.set_title(f"{len(POOL_IDX)} stable sessions pooled", loc="left", color=P["muted"],
             fontsize=12 * FONT_SCALE, pad=4)
ax.set_box_aspect(1); tw.set_box_aspect(1)
dw.label_panel(ax, "a", color=P["fg"])

sc = None
for ax, di in zip(axes[1:], TRAJ_DAYS):
    X, tr = binned(MOUSE, di)
    for p in trial_paths(X, tr, N_TRIALS, SEED):
        s = np.linspace(0, 1, len(p))
        ax.plot(p[:, 0], p[:, 1], color=P["single"], lw=0.6, alpha=0.6, zorder=0)
        sc = ax.scatter(p[:, 0], p[:, 1], c=s, cmap=st.day_cmap(cmap="viridis"), vmin=0, vmax=1,
                        s=5, lw=0, alpha=0.9, zorder=2)
        ax.scatter(p[0, 0], p[0, 1], marker="s", facecolor="none", edgecolor=P["fg"],
                   s=26, lw=1.1, zorder=4)
        ax.scatter(p[-1, 0], p[-1, 1], marker="o", facecolor="none", edgecolor=P["fg"],
                   s=34, lw=1.1, zorder=4)
    ax.set_title(f"day {rel_day(MOUSE, di)}", loc="left", color=P["fg"],
                 fontsize=13 * FONT_SCALE, pad=4)
    ax.set_xlabel(f"PC{PCS[0] + 1}  ({VRAT[PCS[0]]:.1%})")
# one square frame with the SAME span on both axes, shared by every trajectory panel
xl, yl = square_limits(axes[1:], force=LIM)
yl = [-10, 10]
for a in axes[1:]:
    a.set_xlim(*xl); a.set_ylim(*yl); a.set_box_aspect(1)
    a.tick_params(labelleft=(a is axes[1]))
axes[1].set_ylabel(f"PC{PCS[1] + 1}  ({VRAT[PCS[1]]:.1%})")
dw.legend_from(axes[1], [("trial start", "marker", P["fg"], dict(marker="s", ms=6, mfc="none")),
                         ("trial end", "marker", P["fg"], dict(marker="o", ms=7, mfc="none"))],
               loc="upper left", fontsize=10 * FONT_SCALE)
dw.label_panel(axes[1], "b", color=P["fg"])
cb = fig.colorbar(sc, ax=axes[-1], fraction=0.05, pad=0.03)
cb.set_ticks([0, 1]); cb.set_ticklabels(["onset", "target"])
cb.set_label("time in trial", fontsize=11 * FONT_SCALE)
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()
print(f"{MOUSE}: pooled space from {len(POOL_IDX)} stable sessions, {len(MU)} neurons | "
      f"PC1 {VRAT[0]:.1%}, PC1-2 {CUM[1]:.1%}, {DIM80} PCs to reach 80 %, "
      f"participation ratio {PR_POOL:.0f}")
print(f"the two components drawn hold {VRAT[list(PCS)].sum():.1%} of the variance — "
      f"a view, not a summary")
for di in TRAJ_DAYS:
    X, tr = binned(MOUSE, di)
    print(f"   day {rel_day(MOUSE, di)}: {int(np.sum(tr.success == 1))} hits, "
          f"{N_TRIALS} drawn")
```

### 4.2 · One mean trajectory per session

The same idea as §4.1 with the single trials averaged away: **one mean trajectory over that
session's hits**, dark → bright with session number, in three PC pairs (PC1 against PC2, PC3
and PC4). Square = trial start, circle = trial end, as everywhere.

* **Top row, stable control** — how the intuitive axis consolidates over the twenty sessions
  before any rotation.
* **Bottom row, after the rotation** — the mean hit trajectory on the **new** axis
  (factor 3) for each of the fifteen remapping sessions, in the remapping pool's own space.
  The two rows are separate spaces, so PC numbers do not transfer between them.

The arrows are the **decoder axis as estimated from the data**: the least-squares direction
that predicts the rig's recovered $M$ from the binned population ($R^2 \approx 0.9$). It has to
be estimated rather than read from the packet, because the packet's stored weights are a
truncated subset — only 378 of 468 neurons carry one and roughly 62 % of the factor's squared
norm survives. The printout checks the estimate against the stored weights: it agrees at a
cosine of ~0.6–0.75, which is about the ceiling truncation allows ($\sqrt{0.62} = 0.79$).

**One arrow per condition: the across-day mean direction.** Each session's axis is
estimated separately and normalised; the arrow is the normalised mean of those unit vectors,
projected into the plane being drawn. It is **not** per-session and it is **not** drawn at its
true length — only a quarter to a third of either axis lies in any one PC plane, so a
true-length arrow would be invisible. The in-plane fraction is printed instead, together with
how tightly the sessions agree with the mean (|cos| 0.76–0.91 for the stable set, median 0.86;
0.69–0.89 on the remapping blocks), so you can see the average is representative rather than a
cancellation. The bottom row carries both axes, which is the figure that says how far apart
the problem the animal solved twice actually is.

## 5 · Rotating the decoder

Every remapping session of `mouse_0403L` has the same three-block shape and nothing cues the
animal that anything has changed:

| block | axis | when | what it tests |
|---|---|---|---|
| 1 | intuitive (`factor_to_use` 2) | first ~11 min | that control is already stable |
| 2 | rotated (`factor_to_use` 3) | ~11 min → ~50–70 min | whether a new axis can be found |
| 3 | washout, back to intuitive | the last ~6–10 min | whether the old solution survived |

`mouse_0322R` has **no** remapping sessions, so this section is `mouse_0403L` only. The epoch
cell below defines the blocks every figure here compares, so they cannot drift apart.

**The rotation time** is the end of the last intuitive-axis trial — the new block starts the
instant the old one stops — which is `na.remapping_start_frame`. It lands at 10.6–11.1 min on
every one of the 15 sessions. Each block carries **its own reward threshold** (1.9 intuitive,
0.9 rotated), and the tone law is anchored to the threshold, so the target is 8777 Hz in both
blocks and a single reward line serves a whole row. In `CURSOR_REP = "brain"` units that same
line is a step at the block boundary, which is why it is drawn per frame.

**What to expect, having seen §3:** the effect is in the **first exposure**. The figures below
show the first `N_SESSIONS` sessions by default; set `DAYS_I` to see all fifteen.

### 5.1 · Every remapping session, aligned to the rotation

```python
# The epochs every remapping figure compares, and the one colour family they all use: the
# decoder axis that was live. `intuitive` and `new` are the two the summary figures contrast;
# `washout` is kept for the one figure that is about it (5.3) and for 5.4's panel set.
EPOCHS = ["intuitive", "new"]
ECOL   = {"intuitive": C_INT, "new": C_NEW, "washout": C_INT}
ELS    = {"intuitive": "-", "new": "-", "washout": ":"}
EMK    = {"intuitive": "o", "new": "^", "washout": "s"}


def epoch_trials(S):
    """Boolean masks over one session's trial table, one per epoch.

    Blocks come from `factor_to_use` alone (`na.segment_factor_blocks`), never from the
    decoder weights: block 0 is the intuitive axis, block 1 the rotated one, block 2 the
    washout back onto the intuitive axis."""
    bl, idx, out = S.blocks, np.arange(len(S.trials)), {}
    if len(bl) >= 1:
        out["intuitive"] = (idx >= bl[0].first_trial) & (idx <= bl[0].last_trial)
    if len(bl) >= 2:
        out["new"] = (idx >= bl[1].first_trial) & (idx <= bl[1].last_trial)
    if len(bl) >= 3:
        out["washout"] = (idx >= bl[2].first_trial) & (idx <= bl[2].last_trial)
    return out


def block_window(S, k, t0_min, t1_min):
    """Trials of block `k` that START between t0 and t1 minutes after that block began."""
    bl = S.blocks[k]
    idx = np.arange(len(S.trials))
    inb = (idx >= bl.first_trial) & (idx <= bl.last_trial)
    t = (S.trials.start - bl.start_frame) / SR / 60.0
    return inb & (t >= t0_min) & (t < t1_min)


def block_end(S, k):
    """End frame of block k, falling back to the end of the logger where the last trial of a
    session has no recorded end (one session of mouse_0403L)."""
    if k >= len(S.blocks) or not np.isfinite(S.blocks[k].end_frame):
        return float(S.nf)
    return float(S.blocks[k].end_frame)
```

### 5.1b · What the *other* axis was doing — the other-axis simulation

The trace above is the only thing the animal could hear. The interesting question is what the
**inactive** axis was reading at the same moment: during intuitive control, how close was the
population to satisfying the *new* decoder; after the rotation, was it still driving the *old*
one? The animal never heard this and was never rewarded for it. It is a simulation.

The decoder is, exactly (BWAIN / `bmi_idap.Decoder_angle_magnitude`, verified against the
logger):

$$M = X F,\qquad CS_k = \frac{M_k}{\lVert M\rVert},\qquad D_k = |CS_k|^{p} M_k
= \frac{M_k^{3}}{\lVert M\rVert^{2}},\qquad f = F_0\,2^{\,1.5669\, D_k/\theta_k}$$

with $p = 2$ and $\lVert M\rVert$ the norm of the **whole** $K$-dimensional factor-score vector
— which is why the animal can raise the cursor by *suppressing* the other factors, and why the
same $\lVert M\rVert$ serves both axes. Three of these four quantities are not estimated at all:

* $CS$ is logged (`decoder_angles`) and $D$ is logged (`cursor_brain_raw`), so
  $M_{\text{live}} = D/CS^{2}$ and $\lVert M\rVert = D/CS^{3}$ are recovered to machine
  precision. $|CS| < 0.05$ makes that $0/0$ — 7 % of in-session frames, in runs of a **median 1
  and at most 7 frames** — and those gaps are linearly interpolated.
* $\theta_k$ is read per factor from `cursor_threshold_reward` (1.9 intuitive, 0.9 rotated).
* The tone law is exact.

The one estimated quantity is $M_{\text{other}}$: a ridge regression from the population onto
the rig's own $M$, fitted **inside the block where that factor actually was the cursor**, with
the established preprocessing — `spks`, per-neuron z-score over the session, causal exponential
filter with $\tau = 4$ frames.

**The validation is free on a remapping day.** The washout runs the intuitive axis again, so
fitting on block 0 and simulating block 2 is exactly the operation the simulation performs, on
data it never saw, with the answer logged. The cell prints it for all fifteen sessions.

**What it shows.** On the first remapping session the simulated intuitive axis keeps reading
high for minutes after the rotation — it is above the target *more often than the cursor the
animal could actually hear*, which happens on only two of the fifteen sessions. By the third
session the simulated old axis is flat at baseline while the real cursor works. The
printed table is the whole series.

**What that validation says, and the honest limit.** The linear readout transfers well
(r ≈ 0.92–0.96 on held-out data) and the tone follows at r ≈ 0.72–0.85. But $D$ is *cubic* in
$M$, so the gate amplifies whatever error is left **upward**: the simulated trace crosses the
reward threshold more often than the real one does. Read it for **where it sits**, not for how
many rewards the animal would have collected.

**It is not floored at 2963 Hz.** On held-out data the simulated cursor spends as much time
*below* zero as the real one does (the check cell prints both). What looks like a floor is the
linear Hz axis: everything from 2963 Hz down to the 1000 Hz saturation occupies the bottom
eighth of a linear panel. `"hz"` is therefore drawn on a log axis, and `CURSOR_REP = "brain"`
or `"theta"` shows the simulation with zero in the middle and no saturation at all — that is
the representation to judge it in. The simulated trace is also drawn **on top of** the real
one, because underneath it the dense baseline of the real cursor hides it.

```python
# ===== SETTINGS =====
SIM_TAU     = 4.0     # frames: the causal exponential filter standing in for the rig's own
                      # calcium-to-readout smoothing (the validated value, 3-6 all work)
SIM_RIDGE   = 0.1     # ridge penalty in units of n samples; 0 = plain least squares
SIM_CS_MIN  = 0.05    # |CS| below which M = D/CS^2 is 0/0, so ||M|| is interpolated instead
RECOMPUTE_SIM = False
# ====================
# ---------------------------------------------------------------- other-axis simulation
# What the cursor WOULD have been if the OTHER factor had been the decoder direction. The
# animal never heard this; it is the rig's own algebra run on a second axis.
#
# The decoder is, exactly (BWAIN / bmi_idap, verified against the logger):
#     M        = X . F              scores on ALL K factors
#     CS_k     = M_k / ||M||        cosine with factor k, in K-dim factor-score space
#     D_k      = |CS_k|^p . M_k     p = 2, so D_k = M_k^3 / ||M||^2
#     f        = F0 * 2 ** (SLOPE * D_k / theta_k)
# The logger gives CS and D for the LIVE factor, so M_live = D/CS^2 and ||M|| = D/CS^3 are
# recovered to machine precision -- and ||M|| is shared by every factor, which is what makes
# the simulation possible at all. The only unknown is M_other, which is estimated by ridge
# regression from the population onto the rig's own M inside the block where that factor WAS
# live. |CS| < SIM_CS_MIN masks ||M|| (7 % of in-session frames, runs of a median 1 and at
# most 7 frames), and those gaps are linearly interpolated.
#
# The cache stores **D**, which is representation-free, so switching CURSOR_REP costs nothing.


def _causal_exp(X, tau):
    """Causal exponential filter along time, in float32."""
    if tau <= 0:
        return X
    a = float(np.exp(-1.0 / tau))
    return lfilter(np.array([1.0 - a], np.float32), np.array([1.0, -a], np.float32), X, axis=1)


def _ridge(A, y, lam, chunk=20000):
    """Ridge fit with float64 normal equations accumulated in chunks (A may be float32)."""
    n, p = A.shape
    cm = np.zeros(p)
    for s in range(0, n, chunk):
        cm += A[s:s + chunk].astype(np.float64).sum(axis=0)
    cm /= n
    ym = float(np.mean(y))
    G = np.zeros((p, p)); h = np.zeros(p)
    for s in range(0, n, chunk):
        Ac = A[s:s + chunk].astype(np.float64) - cm
        G += Ac.T @ Ac
        h += Ac.T @ (y[s:s + chunk] - ym)
    beta = np.linalg.solve(G + lam * n * np.eye(p), h)
    return beta, ym - cm @ beta


def _fill_gaps(y):
    y = np.asarray(y, float); ok = np.isfinite(y)
    if ok.sum() < 2:
        return y
    out = y.copy(); t = np.arange(len(y))
    out[~ok] = np.interp(t[~ok], t[ok], y[ok])
    return out


_SIM = {}


def simulate_other_axis(mouse, day_i, recompute=False):
    """Per-frame decoder value D for the live axis (re-simulated) and for the OTHER axis."""
    key = (mouse, day_i)
    if key in _SIM:
        return _SIM[key]
    S = session(mouse, day_i)
    tag = f"sim_{mouse}_{S.day}_tau{SIM_TAU:g}_lam{SIM_RIDGE:g}_Dfill"
    fp = CACHE / f"{tag}.npz"
    if fp.exists() and not (recompute or RECOMPUTE_SIM):
        z = np.load(fp, allow_pickle=True)
        out = SimpleNamespace(**{k: z[k] for k in z.files})
        out.val = out.val.item()
        _SIM[key] = out
        return out
    dec = ca.get_decoder_df(R[mouse], S.day)
    D = np.asarray(dec["cursor_brain_raw"], float)[:S.nf]
    CS = np.asarray(dec["decoder_angles"], float)[:S.nf]
    good = np.isfinite(D) & np.isfinite(CS) & (np.abs(CS) > SIM_CS_MIN)
    M_live = np.where(good, D / CS ** 2, np.nan)
    normM = _fill_gaps(np.where(good, D / CS ** 3, np.nan))

    X = na.load_neural(R[mouse], S.day).astype(np.float32)[:, :S.nf]
    X -= X.mean(axis=1, keepdims=True)
    sd = X.std(axis=1, keepdims=True); sd[sd == 0] = 1
    X /= sd
    X = _causal_exp(X, SIM_TAU)                         # (neurons, frames), float32
    XT = np.ascontiguousarray(X.T)                      # (frames, neurons) for the fits
    del X

    # which factor was live on each frame, and which one is therefore "the other"
    fac = np.full(S.nf, np.nan)
    for b in S.blocks:
        e = b.end_frame if np.isfinite(b.end_frame) else S.nf
        fac[int(b.start_frame):int(min(e, S.nf))] = b.factor
    present = sorted({float(b.factor) for b in S.blocks})
    other = {f: [g for g in present if g != f][0] if len(present) > 1 else f for f in present}

    # A block runs from its FIRST true BMI trial to its LAST, so the rig's ~30 s changeover
    # between two blocks carries no factor at all, and so did the simulation. Carry every
    # unlabelled frame BACKWARD from the next block that has a label (and the tail forward
    # from the last one), which also covers the lead-in before the session's first trial.
    # `fac` itself stays strict and is what the fits below use: a changeover frame must never
    # join a factor's training set, because the rig's own decoder may still be the old one.
    fac_f = fac.copy()
    _k = np.where(np.isfinite(fac))[0]
    if len(_k):
        fac_f = fac[_k[np.clip(np.searchsorted(_k, np.arange(S.nf), "left"), 0, len(_k) - 1)]]

    base = S.in_exp & good & np.isfinite(M_live)
    Mhat = {}
    for f in present:
        tr_m = base & (fac == f)
        beta, c = _ridge(XT[tr_m], M_live[tr_m], SIM_RIDGE)
        Mhat[f] = XT @ beta + c

    fac_other = np.array([other.get(v, np.nan) if np.isfinite(v) else np.nan
                          for v in fac_f])
    th_live = np.array([S.thr_by_f.get(v, np.nan) if np.isfinite(v) else np.nan
                        for v in fac_f])
    th_other = np.array([S.thr_by_f.get(v, np.nan) if np.isfinite(v) else np.nan
                         for v in fac_other])
    M_o = np.full(S.nf, np.nan); M_l = np.full(S.nf, np.nan)
    for f in present:
        M_l[fac_f == f] = Mhat[f][fac_f == f]
        M_o[fac_other == f] = Mhat[f][fac_other == f]
    D_other = M_o ** 3 / normM ** 2
    D_live = M_l ** 3 / normM ** 2
    D_other[~S.in_exp] = np.nan; D_live[~S.in_exp] = np.nan

    # ---- held-out check: fit the intuitive axis on block 0, simulate the washout ----
    val = {}
    if len(S.blocks) >= 3 and S.blocks[0].factor == S.blocks[2].factor:
        f0 = float(S.blocks[0].factor); th = S.thr_by_f[f0]
        m_fit = base & (fac == f0) & (np.arange(S.nf) < block_end(S, 0))
        m_te = base & (fac == f0) & (np.arange(S.nf) >= S.blocks[2].start_frame) & ~S.ovr
        if m_fit.sum() > 500 and m_te.sum() > 500:
            beta, c = _ridge(XT[m_fit], M_live[m_fit], SIM_RIDGE)
            Mp = XT[m_te] @ beta + c
            Dp = Mp ** 3 / normM[m_te] ** 2
            hp = d_to_hz(Dp, th)
            val = dict(n=int(m_te.sum()),
                       r_M=float(np.corrcoef(Mp, M_live[m_te])[0, 1]),
                       r_D=float(np.corrcoef(Dp, D[m_te])[0, 1]),
                       r_hz=float(np.corrcoef(hp, S.hz_raw[m_te])[0, 1]),
                       lo_sim=float(np.mean(Dp < 0)), lo_real=float(np.mean(D[m_te] < 0)),
                       frac_sim=float(np.mean(hp >= HZ_REWARD)),
                       frac_real=float(np.mean(S.hz_raw[m_te] >= HZ_REWARD)))
    out = SimpleNamespace(D_other=D_other.astype(np.float32), D_live=D_live.astype(np.float32),
                          th_other=th_other.astype(np.float32),
                          th_live=th_live.astype(np.float32),
                          M_other=M_o.astype(np.float32), M_live=M_live.astype(np.float32),
                          normM=normM.astype(np.float32),
                          fac=fac_f.astype(np.float32),
                          fac_block=fac.astype(np.float32),   # NaN across the changeovers
                          fac_other=fac_other.astype(np.float32),
                          Mhat=np.array([Mhat[f] for f in present], np.float32),
                          present=np.array(present, float),
                          val=val)
    np.savez_compressed(fp, **vars(out))
    _SIM[key] = out
    return out


def sim_cursor(sim, which="other", smooth=True):
    """The simulated cursor in the ACTIVE representation. `which` is "other" or "live"."""
    D = np.asarray(getattr(sim, f"D_{which}"), float)
    th = np.asarray(getattr(sim, f"th_{which}"), float)
    y = d_to_rep(D, th)
    return smooth_nan(y, SMOOTH_SEC * SR) if smooth else y


_sims = [simulate_other_axis(MOUSE, int(i)) for i in REMAP_I[MOUSE]]
_v = [s.val for s in _sims if s.val]
_gap = [float(np.sum(~np.isfinite(s.fac_block) & session(MOUSE, int(i)).in_exp)) / SR
        for s, i in zip(_sims, REMAP_I[MOUSE])]
print(f"held-out check on {len(_v)} remapping sessions (fit the intuitive axis on block 0, "
      f"simulate it through the full gate and tone law in the washout):")
print(f"   linear readout M : r = {np.min([v['r_M'] for v in _v]):.2f}"
      f"–{np.max([v['r_M'] for v in _v]):.2f}  (median "
      f"{np.median([v['r_M'] for v in _v]):.2f})")
print(f"   gated cursor D   : r = {np.min([v['r_D'] for v in _v]):.2f}"
      f"–{np.max([v['r_D'] for v in _v]):.2f}")
print(f"   tone (Hz)        : r = {np.min([v['r_hz'] for v in _v]):.2f}"
      f"–{np.max([v['r_hz'] for v in _v]):.2f}")
print(f"   frames with D below 0 (the half a linear Hz axis hides): simulated "
      f"{np.median([v['lo_sim'] for v in _v]):.1%} vs real "
      f"{np.median([v['lo_real'] for v in _v]):.1%} (medians) — the simulated cursor is "
      f"NOT floored at the base tone; it spends as much time below zero as the real one")
print(f"   frames at or above the reward threshold: simulated "
      f"{np.median([v['frac_sim'] for v in _v]):.2%} vs real "
      f"{np.median([v['frac_real'] for v in _v]):.2%} (medians) — the cube in the gate "
      f"amplifies the estimate's error upward, so read the simulated trace for WHERE it "
      f"sits, not for how often it would have been rewarded")
print(f"   in-session frames with no block label (the changeover between blocks, plus the "
      f"lead-in before the first trial): {np.median(_gap):.0f} s per session, range "
      f"{np.min(_gap):.0f}–{np.max(_gap):.0f} s. They take the NEXT block's factor so the "
      f"simulation is defined everywhere, and they are kept out of every fit.")
```

### 5.2 · What the rotation costs

The same two measures as §3.1 and §3.2, now within a session rather than across days, split by
block: the whole intuitive block against the whole rotated block. No early/late split and no
washout — the question here is only what the rotation itself costs, and §5.3 is where the
washout belongs.

**The cost is entirely in the first exposure.** On the first remapping session the hit rate
falls from 100 % to 74 % and the median time to target goes from about 1 s to about 6 s. On
every session after it the animal is at or near the intuitive-block level, and pooled over all
fifteen sessions a paired test finds no effect at all — the cell prints both. An animal that
has met one rotation solves the next one immediately.

One thing not to over-read: the intuitive block is **not** a matched control for time on task —
it is always first, and §3.4 shows the reward rate falling over the hour that follows it.

```python
# ===== SETTINGS =====
DAYS_I    = None        # None = every remapping session of MOUSE
HIT_YLIM  = (0.5, 1.04)
T_YLIM    = (0, None)
FIGSIZE   = (13.5, 5.2)
TITLE     = None
FIGNAME   = "fig10_remap_block_performance"
# ====================
# One line per axis, one point per session: the whole intuitive block against the whole
# rotated block. No early/late split and no washout -- the question here is only what the
# rotation itself costs, and 5.3 is where the washout belongs.
RD = [int(i) for i in (REMAP_I[MOUSE] if DAYS_I is None else DAYS_I)]
rows = []
for di in RD:
    S = session(MOUSE, di)
    dur = (S.trials.end - S.trials.start) / SR
    for e, m in epoch_trials(S).items():
        if e not in EPOCHS or not m.any():
            continue
        hit = S.trials.success[m] == 1
        rows.append(dict(i=di, rel=rel_day(MOUSE, di), epoch=e,
                         n=int(m.sum()), hit_rate=float(np.mean(hit)),
                         t_hit=float(np.nanmedian(dur[m][hit])) if hit.any() else np.nan))
EP = pd.DataFrame(rows)

fig, axes = plt.subplots(1, 2, figsize=FIGSIZE, constrained_layout=True)
for ax, col, lab, ylim in [(axes[0], "hit_rate", "hit rate", HIT_YLIM),
                           (axes[1], "t_hit", "median time to target (s)", T_YLIM)]:
    for e in EPOCHS:
        s = EP[EP["epoch"] == e].sort_values("rel")
        if not len(s):
            continue
        ax.plot(s["rel"], s[col], color=ECOL[e], ls=ELS[e], marker=EMK[e], ms=6, lw=2.2)
    ax.set_xlabel("session (days after day 0)"); ax.set_ylabel(lab)
    ax.set_ylim(*ylim)
# the whole effect lives on the first exposure, so say so on the figure
first = EP[EP["rel"] == EP["rel"].min()].set_index("epoch")
axes[1].annotate("first exposure\nto the rotation",
                 xy=(first.loc["new", "rel"], first.loc["new", "t_hit"]),
                 xytext=(18, -6), textcoords="offset points", ha="left", va="top",
                 color=C_NEW, fontsize=11 * FONT_SCALE,
                 arrowprops=dict(arrowstyle="-", color=C_NEW, lw=1.2))
dw.legend_from(axes[0], [(f"{e} axis", "line", ECOL[e], dict(lw=2.4, ls=ELS[e],
                                                             marker=EMK[e], ms=6))
                         for e in EPOCHS],
               loc="lower right", fontsize=11 * FONT_SCALE)
dw.label_panel(axes[0], "a", color=P["fg"]); dw.label_panel(axes[1], "b", color=P["fg"])
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()

piv = EP.pivot(index="rel", columns="epoch", values=["hit_rate", "t_hit"])
print(EP.groupby("epoch", sort=False)[["n", "hit_rate", "t_hit"]]
        .agg(["mean", "min", "max"]).round(2).to_string())
a, b = piv["hit_rate"]["intuitive"], piv["hit_rate"]["new"]
ta, tb = piv["t_hit"]["intuitive"], piv["t_hit"]["new"]
print(f"\nFIRST remapping session (day {piv.index[0]}): hit rate {a.iloc[0]:.0%} → "
      f"{b.iloc[0]:.0%}, time to target {ta.iloc[0]:.1f} → {tb.iloc[0]:.1f} s "
      f"({tb.iloc[0] / ta.iloc[0]:.1f}x slower)")
print(f"every later remapping session:   hit rate {a.iloc[1:].mean():.0%} → "
      f"{b.iloc[1:].mean():.0%}, time to target {ta.iloc[1:].mean():.1f} → "
      f"{tb.iloc[1:].mean():.1f} s")
print(f"pooled over all {len(piv)} sessions the rotation is not detectable: hit rate "
      f"p = {stats.wilcoxon(a, b).pvalue:.2g}, time to target p = {stats.wilcoxon(ta, tb).pvalue:.2g} "
      f"(Wilcoxon signed-rank) — the cost is in the first exposure, not in the rotation")
```

### 5.3 · The washout

The cleanest version of the claim is inside a single session: the axis is rotated away, the
animal struggles for the better part of an hour, and then the original decoder is restored
without warning and control is back on the very next trials.

This is plotted **trial by trial** rather than frame by frame. The two quantities the claim is
about — how long the animal needs to reach the target, and how often it gets there — are one
number per trial each, and 80 minutes of 30 Hz cursor on a single row reads as noise; §5.1
shows the raw trace around the rotation itself. Each dot is a hit and each cross a miss; the **white line is the rolling median time to target** (left axis, same units as
the dots) and the **cyan line the rolling hit rate** (right axis), both over the same
`ROLL`-trial window. Blocks are shaded by which axis was live, with each block's hit rate and
median time to target printed above it.

On the first remapping session the washout is a **complete** return: the same 100 % hit rate as
the opening block and, if anything, a slightly shorter time to target. The second session is
noisier — by then its washout is at minute 71 of an 80-minute session — which §3.4 explains
better than any claim about the axis.

Note what this does *not* show: the animal does not master the new axis within that first
session (the rotated block ends no better than it started). The improvement happens **between**
sessions, which is why §5.2's curves are flat from the second remapping session on.

```python
# ===== SETTINGS =====
N_DAYS   = 3           # "the first day or two"
ROLL     = 15           # trials in the rolling window, shared by both rolling measures
C_RATE   = "#00c2d1"    # the rolling HIT RATE line. Not C_HIT: that is the notebook-wide
                        # colour of a hit trial, and rebinding it here would leak into
                        # every cell that runs after this one.
C_DUR    = "#e8e8e8"    # rolling median time to target, in the same units as the dots
YLIM     = (0, 24)
FIGSIZE  = (14.0, 6.8)
TITLE    = None
FIGNAME  = "fig11_washout"
# ====================
# Trial by trial, not frame by frame: an 80 min tone trace at 30 Hz is 140 000 points on one
# row and reads as noise. The two things the claim is about -- how long the animal needs to
# reach the target, and how often it gets there -- are one number per trial each, and the
# same ROLL-trial window carries both.
RD = [int(i) for i in REMAP_I[MOUSE][:N_DAYS]] 
fig, axes = plt.subplots(len(RD), 1, figsize=FIGSIZE, sharex=True, sharey=True,
                         constrained_layout=True)
for ax, di in zip(np.atleast_1d(axes), RD):
    S = session(MOUSE, di)
    t = S.trials.start / SR / 60.0
    d = (S.trials.end - S.trials.start) / SR
    hit = S.trials.success == 1
    base_f = S.trials.factor[0]
    for k, b in enumerate(S.blocks):
        ax.axvspan(b.start_frame / SR / 60.0, block_end(S, k) / SR / 60.0, lw=0, zorder=0,
                   alpha=0.12, color=(C_INT if b.factor == base_f else C_NEW))
    #ax.scatter(t[hit], d[hit], s=14, color=P["muted"], lw=0, alpha=0.8, zorder=2)
    #ax.scatter(t[~hit], np.full(int((~hit).sum()), 20.4), s=20, marker="x",
    #           color=P["muted"], lw=1.1, alpha=0.7, zorder=2)
    # rolling median time to target, over hits only (a miss is censored at the 20 s timeout)
    dh = pd.Series(np.where(hit, d, np.nan))
    ax.plot(t, dh.rolling(ROLL, center=True, min_periods=5).median(), color=C_DUR, lw=2.4,
            zorder=4)
    tw = ax.twinx(); tw.spines["right"].set_visible(True)
    roll = pd.Series(hit.astype(float)).rolling(ROLL, center=True, min_periods=5).mean()
    tw.plot(t, roll, color=C_RATE, lw=2.4, zorder=4)
    tw.set_ylim(0, 1.30); tw.set_yticks([0, 0.25, 0.5, 0.75, 1.0])
    tw.set_ylabel("hit rate", color=C_RATE, fontsize=12 * FONT_SCALE)
    tw.tick_params(axis="y", colors=C_RATE)
    for k, b in enumerate(S.blocks):
        sl = slice(b.first_trial, b.last_trial + 1)
        h = S.trials.success[sl] == 1
        ax.text((b.start_frame + block_end(S, k)) / 2 / SR / 60.0, YLIM[1] * 0.99,
                f"{np.mean(h):.0%}   {np.nanmedian(d[sl][h]):.1f} s", ha="center", va="top",
                fontsize=11.5 * FONT_SCALE,
                color=(C_INT if b.factor == base_f else C_NEW))
    ax.set_xlim(0, S.nf / SR / 60.0); ax.set_ylim(*YLIM)
    ax.set_ylabel(f"session {rel_day(MOUSE, di)}\ntime to target (s)", color=C_DUR,
                  fontsize=12 * FONT_SCALE)
    ax.tick_params(axis="y", colors=C_DUR)
np.atleast_1d(axes)[-1].set_xlabel("time in the session (min)")
dw.legend_from(np.atleast_1d(axes)[0],
               [("intuitive axis", "patch", C_INT, dict(alpha=0.30)),
                ("rotated axis", "patch", C_NEW, dict(alpha=0.30)),
                ("one hit", "marker", P["muted"], dict(marker="o", ms=6)),
                ("a miss (20 s timeout)", "marker", P["muted"], dict(marker="x", ms=7)),
                (f"median time to target, {ROLL}-trial window", "line", C_DUR, dict(lw=2.4)),
                (f"hit rate, {ROLL}-trial window", "line", C_RATE, dict(lw=2.4))],
               loc="lower center", bbox_to_anchor=(0.5, 1.03), ncol=3, fontsize=11 * FONT_SCALE)
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()
for di in RD:
    S = session(MOUSE, di)
    dur = (S.trials.end - S.trials.start) / SR
    out = []
    for k, b in enumerate(S.blocks):
        sl = slice(b.first_trial, b.last_trial + 1)
        h = S.trials.success[sl] == 1
        out.append(f"block {k} (factor {b.factor:g}, {b.n_trials} trials, "
                   f"{b.start_frame / SR / 60:.0f}–{block_end(S, k) / SR / 60:.0f} min): "
                   f"{np.mean(h):.0%}, {np.nanmedian(dur[sl][h]):.1f} s")
    print(f"session {rel_day(MOUSE, di)}\n   " + "\n   ".join(out))
```

then four windows of the same session in one square frame — the intuitive block, the **first

Trials are sampled at random (fixed seed) so the panels stay readable, and **both hits and

The comparison the middle two panels are for: early in the rotated block the paths wander and

```python
# ===== SETTINGS =====
REMAP_DAY_I = None      # None = the first remapping session
N_TRIALS    = 12        # trials drawn per panel, sampled at random with a fixed seed
PCS         = (0, 1)    # which two components of that day's own PCA to draw
N_SPEC      = 30
LIM         = None      # (lo, hi) on BOTH axes of every PCA panel; None = auto
SEED        = 2
FIGSIZE     = (16.0, 4.6)
TITLE       = None
FIGNAME     = "fig12_remap_day_pca"
# ====================
RDI = int(REMAP_I[MOUSE][0] if REMAP_DAY_I is None else REMAP_DAY_I)
Xr, trr = binned(MOUSE, RDI)
_valr, _vecr, _vratr = eig_space(np.cov(Xr))
Sr = session(MOUSE, RDI)
EM = epoch_trials(Sr)
# a is the spectrum; b-e are four windows of the session, in that session's own PCA space
PANELS = [("intuitive axis", EM["intuitive"], C_INT),
          ("new axis, first 15 min", block_window(Sr, 1, 0, 15), C_NEW),
          ("new axis, 45–60 min", block_window(Sr, 1, 45, 60), C_NEW),
          ("washout (intuitive again)", EM.get("washout", np.zeros(len(Sr.trials), bool)), C_INT)]
VKr = _vecr[:, list(PCS)]
MUr_day = Xr.mean(axis=1)

fig, axes = plt.subplots(1, 1 + len(PANELS), figsize=FIGSIZE, constrained_layout=True)
ax = axes[0]
k = np.arange(1, N_SPEC + 1)
ax.bar(k, _vratr[:N_SPEC] * 100, color=P["muted"], width=0.8)
tw = ax.twinx(); tw.spines["right"].set_visible(True)
tw.plot(k, np.cumsum(_vratr)[:N_SPEC] * 100, color=P["fg"], lw=2.0)
tw.set_ylabel("cumulative (%)", fontsize=12 * FONT_SCALE); tw.set_ylim(0, 100)
ax.set_xlabel("component"); ax.set_ylabel("variance explained (%)")
ax.set_title(f"session {rel_day(MOUSE, RDI)}", loc="left", color=P["muted"],
             fontsize=12 * FONT_SCALE, pad=4)
ax.set_box_aspect(1); tw.set_box_aspect(1)

rng = np.random.default_rng(SEED)
for ax, (lab, m, col) in zip(axes[1:], PANELS):
    ok = np.flatnonzero(m & np.isfinite(trr.start) & np.isfinite(trr.end))
    if len(ok) > N_TRIALS:
        ok = np.sort(rng.choice(ok, N_TRIALS, replace=False))
    n_hit = 0
    for j in ok:
        a, b = int(trr.start[j]), min(int(trr.end[j]), Xr.shape[1])
        if b - a < 3:
            continue
        p = gaussian_filter1d((Xr[:, a:b].T - MUr_day) @ VKr, SMOOTH_BIN, axis=0,
                              mode="nearest")
        is_hit = trr.success[j] == 1
        n_hit += int(is_hit)
        ax.plot(p[:, 0], p[:, 1], color=col, lw=1.0, alpha=0.55 if is_hit else 0.35,
                ls="-" if is_hit else "--", zorder=2)
        ax.scatter(p[0, 0], p[0, 1], color=col, s=26, marker="s", lw=0, zorder=4)
        if is_hit:
            ax.scatter(p[-1, 0], p[-1, 1], facecolor="none", edgecolor=col, s=46, lw=1.6,
                       marker="o", zorder=5)
        else:
            ax.scatter(p[-1, 0], p[-1, 1], color=col, s=46, lw=1.8, marker="x", zorder=5)
    ax.set_title(f"{lab}\n{n_hit}/{len(ok)} hits drawn", loc="left", color=col,
                 fontsize=11.5 * FONT_SCALE, pad=4)
    ax.set_xlabel(f"PC{PCS[0] + 1}  ({_vratr[PCS[0]]:.1%})")
xl, yl = square_limits(axes[1:], force=LIM)
for a in axes[1:]:
    a.set_xlim(*xl); a.set_ylim(*yl); a.set_box_aspect(1)
    a.tick_params(labelleft=(a is axes[1]))
axes[1].set_ylabel(f"PC{PCS[1] + 1}  ({_vratr[PCS[1]]:.1%})")
dw.legend_from(axes[-1], [("trial start", "marker", P["muted"], dict(marker="s", ms=6)),
                          ("hit", "marker", P["muted"], dict(marker="o", ms=7, mfc="none")),
                          ("miss", "marker", P["muted"], dict(marker="x", ms=7))],
               loc="upper left", fontsize=10 * FONT_SCALE)
for a, letter in zip(axes, "abcde"):
    dw.label_panel(a, letter, color=P["fg"])
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()
print(f"{DAYS[MOUSE][RDI]} (session {rel_day(MOUSE, RDI)}): PC1 {_vratr[0]:.1%}, "
      f"PC1-2 {np.cumsum(_vratr)[1]:.1%}, "
      f"{int(np.searchsorted(np.cumsum(_vratr), 0.8) + 1)} PCs to 80 %, "
      f"participation ratio {_valr.sum() ** 2 / (_valr ** 2).sum():.0f}")
for lab, m, _c in PANELS:
    h = Sr.trials.success[m] == 1
    ends = [((Xr[:, int(trr.start[j]):min(int(trr.end[j]), Xr.shape[1])].T - MUr_day) @ VKr)[-1, 0]
            for j in np.flatnonzero(m & (trr.success == 1))
            if min(int(trr.end[j]), Xr.shape[1]) - int(trr.start[j]) >= 3]
    print(f"   {lab}: {int(m.sum())} trials, {np.mean(h):.0%} hits; hits end at "
          f"PC{PCS[0] + 1} = {np.median(ends):+.1f} (median of {len(ends)})")
```

### 5.5 · Three windows of the first remapping session

The same session as §5.4, read as three 50-second windows rather than as a trial average:
**before the rotation** (exploitation, 8.9 min), **early after it** (exploration, 13.9 min) and
**late after it** (the new solution, 54.93 min). The cursor the animal heard is on top; the
population trajectory over the identical window is below, in one PCA fitted on the whole
session. Path colour is the behavioural state — hit trial, miss trial, or the inter-trial
interval drawn thin. The cursor row also carries the other-axis simulation of §5.1b.

Three fixed markers appear in every column, so the three windows can be read against the same
landmarks:

* $T_{int}$ — the mean population state over the last 0.5 s of a **pre-rotation hit**:
  the intuitive solution.
* $T_{new}$ — the same over the **late half** of the rotated block, so it is the new solution
  *after* it has been found.
* The **quiescence baseline** (diamond — a square would read as a trial start) — the mean
  state over frames the task itself calls
  quiescent and that fall outside any trial. This is the state the rig requires before it will
  start a trial, which makes it the right origin; the inter-trial mean is not, because the ITI
  still contains whatever the animal was doing between attempts.

The two arrows are the session's own estimated decoder directions, intuitive and new, drawn from
that baseline. As everywhere in this notebook they show **direction at a fixed length** — the
in-plane fraction is printed — and they are drawn from the baseline rather than from the origin
so they point from where the animal rests toward what each decoder is asking for.

```python
# ===== SETTINGS =====
DAY_I     = None        # None = the first remapping session
WINDOWS   = [("before the rotation\n(exploitation)", 8.90),
             ("early after the rotation\n(exploration)", 13.90),
             ("late after the rotation\n(the new solution)", 54.93)]
DUR_SEC   = 50.0
PCS       = (0, 1)
LIM       = None        # (lo, hi) on BOTH axes of every trajectory panel; None = auto
CUR_YLIM  = None        # None = the active representation's own limits
SHOW_SIM  = True        # overlay the other-axis simulation on the cursor row
TRAJ_LW   = 1.9         # line width during trials
ITI_LW    = 0.7         # line width between them
ENDPOINT_SEC = 0.5      # last N s of a hit that defines a solution state
SHOW_AXES = True        # the two estimated decoder directions, as arrows from the baseline
ARROW_FRAC = 0.42
FIGSIZE   = (14.5, 6.6)
TITLE     = None
FIGNAME   = "fig13_rotation_windows"
# ====================
DI = int(REMAP_I[MOUSE][0] if DAY_I is None else DAY_I)
S = session(MOUSE, DI)
X, trb = binned(MOUSE, DI)
SRB = SR / BIN
MU_D = X.mean(axis=1)
_val, _vec, _vrat = eig_space(np.cov(X))
VK = _vec[:, list(PCS)]
proj = lambda A: (np.atleast_2d(A) - MU_D) @ VK

# the two solution states, exactly as the animations define them: the mean population state
# over the last ENDPOINT_SEC of a hit -- in the pre-rotation block for T_int, and over the
# LATE HALF of the rotated block for T_new, so it reflects the solution once found.
w = max(int(round(ENDPOINT_SEC * SRB)), 1)


def block_endpoints(k, late_half=False):
    b = S.blocks[k]
    idx = [i for i in range(b.first_trial, b.last_trial + 1) if trb.success[i] == 1]
    if late_half:
        idx = idx[len(idx) // 2:]
    out = []
    for i in idx:
        s_, e_ = max(int(round(trb.start[i])), 0), min(int(round(trb.end[i])), X.shape[1])
        if e_ > s_:
            out.append(X[:, max(e_ - w, s_):e_].mean(1))
    return np.array(out) if out else np.empty((0, X.shape[0]))


T_INT = block_endpoints(0).mean(0)
T_NEW = block_endpoints(1, late_half=True).mean(0)

# the baseline the animal returns to: the mean state during QUIESCENCE, i.e. the gate the
# task itself uses to start a trial -- not the inter-trial interval, which still contains
# whatever the animal was doing between attempts.
nb = X.shape[1]
qf = S.quiet[:nb * BIN].reshape(nb, BIN).mean(1) > 0.5
lab_all = na.label_timepoints(trb, nb, sr=SRB)
q_bins = qf & (lab_all == na.STATE_ITI)
BASE = X[:, q_bins].mean(1)

# this day's two estimated decoder directions, from the pooled remapping pass
_j = int(np.flatnonzero(RIDX == DI)[0])
U_INT, U_NEW = AXI[_j], AXN[_j]
_sim = simulate_other_axis(MOUSE, DI) if SHOW_SIM else None

fig = plt.figure(figsize=FIGSIZE, constrained_layout=True)
gs = fig.add_gridspec(2, len(WINDOWS), height_ratios=[1.0, 3.6], hspace=0.06, wspace=0.08)
axC = [fig.add_subplot(gs[0, c]) for c in range(len(WINDOWS))]
axT = [fig.add_subplot(gs[1, c]) for c in range(len(WINDOWS))]
for c, (lab, t0) in enumerate(WINDOWS):
    f0 = int(round(t0 * 60 * SR)); f1 = int(min(f0 + DUR_SEC * SR, S.nf))
    b0, b1 = f0 // BIN, min(int(f1 // BIN), nb)
    ts = (np.arange(f0, f1) - f0) / SR
    # ---- the cursor over this window ----
    a = axC[c]
    dw.shade_trials(a, S.trials.start, S.trials.end, S.trials.success,
                    colors={1.0: C_HIT, 0.0: C_MISS}, alpha=0.18,
                    sr=SR, t0=f0, clip=(f0, f1))
    if SHOW_SIM:
        oth = sim_cursor(_sim, "other")[f0:f1]
        fo = np.asarray(_sim.fac_other, float)[f0:f1]
        for ff, cc in [(F_INT, C_INT), (F_NEW, C_NEW)]:
            a.plot(ts, np.where(fo == ff, oth, np.nan), color=cc, lw=0.8, alpha=0.9, zorder=3)
    a.plot(ts, S.cur[f0:f1], color=P["cursor"], lw=0.9, zorder=2)
    a.plot(ts, S.rew_level[f0:f1], color=P["correct"], lw=1.0, zorder=1)
    a.set_xlim(0, DUR_SEC)
    rep_axis(a, CUR_YLIM, ticks=False, label=False)
    a.set_yticks([])
    a.set_title(lab, loc="center", color=P["fg"], fontsize=12.5 * FONT_SCALE, pad=6)
    a.set_xlabel("time (s)", fontsize=11 * FONT_SCALE, labelpad=1)
    a.tick_params(labelsize=10 * FONT_SCALE)
    if c == 0:
        a.set_ylabel("cursor", fontsize=11 * FONT_SCALE)
    # ---- the population trajectory over the same window ----
    ax = axT[c]
    p = gaussian_filter1d((X[:, b0:b1].T - MU_D) @ VK, SMOOTH_BIN, axis=0, mode="nearest")
    lab_w = lab_all[b0:b1]
    cols = np.where(lab_w[:-1] == na.STATE_CORRECT, C_HIT,
                    np.where(lab_w[:-1] == na.STATE_INCORRECT, C_MISS, P["single"]))
    lws = np.where(lab_w[:-1] == na.STATE_ITI, ITI_LW, TRAJ_LW)
    ax.add_collection(LineCollection(np.stack([p[:-1], p[1:]], axis=1), colors=cols,
                                     linewidths=lws, zorder=2))
    for v, c_ in [(T_INT, C_INT), (T_NEW, C_NEW)]:
        q = proj(v)[0]
        ax.scatter(q[0], q[1], s=170, marker="X", color=c_, edgecolor=P["bg"], lw=0.8, zorder=9)
    qb = proj(BASE)[0]
    ax.scatter(qb[0], qb[1], s=110, marker="D", color=P["fg"], edgecolor=P["bg"], lw=0.8,
               zorder=8)
    ax.set_xlabel(f"PC{PCS[0] + 1}  ({_vrat[PCS[0]]:.1%})")
    if c == 0:
        ax.set_ylabel(f"PC{PCS[1] + 1}  ({_vrat[PCS[1]]:.1%})")
# one square frame on every trajectory panel
xl, yl = square_limits(axT, force=LIM)
lo, hi = xl
for c, ax in enumerate(axT):
    ax.set_xlim(*xl); ax.set_ylim(*yl); ax.set_box_aspect(1)
    ax.tick_params(labelleft=(c == 0))
    if SHOW_AXES:
        # the two decoder directions, drawn from the quiescence baseline at a fixed length:
        # only a few per cent of either axis lies in this plane, so the length is arbitrary
        # and only the direction is meaningful.
        o = proj(BASE)[0]
        for u, c_ in [(U_INT, C_INT), (U_NEW, C_NEW)]:
            v = u @ VK
            v = ARROW_FRAC * (hi - lo) * v / np.linalg.norm(v)
            ax.annotate("", xy=tuple(o + v), xytext=tuple(o), zorder=7,
                        arrowprops=dict(arrowstyle="-|>", color=c_, lw=2.4, alpha=0.95,
                                        shrinkA=0, shrinkB=0, mutation_scale=18))
dw.legend_from(axT[-1],
               [("hit trial", "line", C_HIT, dict(lw=TRAJ_LW)),
                ("miss trial", "line", C_MISS, dict(lw=TRAJ_LW)),
                ("inter-trial interval", "line", P["single"], dict(lw=ITI_LW)),
                ("$T_{int}$ (intuitive solution)", "marker", C_INT, dict(marker="X", ms=9)),
                ("$T_{new}$ (new solution)", "marker", C_NEW, dict(marker="X", ms=9)),
                ("quiescence baseline", "marker", P["fg"], dict(marker="D", ms=7)),
                ("decoder directions", "line", P["muted"], dict(lw=2.2))],
               loc="upper left", bbox_to_anchor=(1.02, 1.0), fontsize=10 * FONT_SCALE)
for a_, letter in zip(axC, "abc"):
    dw.label_panel(a_, letter, color=P["fg"])
if TITLE: fig.suptitle(TITLE)
st.save_fig(fig, FIGNAME); plt.show()
print(f"{S.day} (session {rel_day(MOUSE, DI)}), one PCA on the whole session: "
      f"PC1 {_vrat[0]:.1%}, PC1-2 {_vrat[:2].sum():.1%}; rotation at {S.remap / SR / 60:.2f} min, "
      f"washout at {S.blocks[2].start_frame / SR / 60:.2f} min")
print(f"quiescence baseline from {int(q_bins.sum())} bins "
      f"({q_bins.mean():.0%} of the session; quiescence outside any trial)")
for nm, v in [("T_int", T_INT), ("T_new", T_NEW), ("baseline", BASE)]:
    q = proj(v)[0]
    print(f"   {nm}: PC{PCS[0]+1} = {q[0]:+.1f}, PC{PCS[1]+1} = {q[1]:+.1f}")
print(f"angle between this session's two estimated decoder axes: {ang[_j]:.0f} deg "
      f"(two random directions in {len(MUr)}-D would be {chance:.0f} deg apart)")
print(f"in-plane fraction: intuitive {np.linalg.norm(U_INT @ VK):.0%}, "
      f"new {np.linalg.norm(U_NEW @ VK):.0%} — the arrows show direction only")
```

## 6 · The search, as a movie

The static panels of §5.5 pick three windows. The movie is the same thing without the picking:
one PCA space per session, a fading trail of the recent trajectory coloured by behavioural
state, the two solution states fixed in place, and the cursor scrolling underneath.

Layout: **PC1–PC2** large on the left, **PC2–PC3** and **PC1–PC3** small underneath, and
**two** cursor rows top right — what the animal actually drove, and the **other-axis
simulation** of §5.1b underneath it, in the colour of the axis being simulated. They are on
separate rows because overlaid they are unreadable; the purple trace during the rotated block
is "what the old decoder would have been reading", and the orange trace before the rotation is
"how close the animal already was to the new one". $T_{int}$, $T_{new}$ and the quiescence
baseline are marked in every PC panel, and the panel limits are widened to hold them.
`LEGEND = False` by default — the keys are in §5.5.

**One short movie by default** — the first remapping session, one minute before the rotation
to three minutes after. Put `[0, 5, 11]` back in `VIDEO_DAYS` and widen `PRE_MIN`/`POST_MIN`
for the full set (1st, 6th and 12th sessions: first exposure, mid-series, late). They are
written to `output/wrapup_notebooks/02_learning_and_remapping/videos/`. `STEP_SEC` sets the
playback speed (data-seconds per frame); `RENDER_VIDEOS = False` skips the whole cell.

```python
# ===== SETTINGS =====
RENDER_VIDEOS = True          # if False, the code runs but no video is written
VIDEO_DAYS = [7]               # which remapping sessions, in order: [0, 5, 11] = 1st, 6th, 12th
PRE_MIN    = 5               # minutes of intuitive control before the rotation
POST_MIN   = 20              # minutes of the new axis after it
STEP_SEC   = 0.2               # data-seconds advanced per video frame (lower = slower)
TRAIL_SEC  = 20.0              # how much of the recent path stays drawn
FPS, DPI   = 20, 110
TS_WIN_MIN = 1.5               # width of the scrolling cursor panels
PC_PAIRS_V = [(0, 1), (1, 2), (0, 2)]
LEGEND     = False             # the panels are busy enough; True puts the keys back
CUR_HSPACE = 0.06              # gap between the two cursor rows (they share the x axis)
N_ENDMARK  = 10                # how many recent trial-end circles stay on the PC panels
ENDMARK_A  = 0.7               # ...and how solid they are
FIGSIZE_V  = (16.0, 9.0)
# ====================
# One short movie by default. Widen PRE_MIN/POST_MIN and put [0, 5, 11] back in VIDEO_DAYS
# for the full set; at STEP_SEC = 0.5 a 15 min window is ~1800 frames and a few minutes each.
import imageio_ffmpeg
plt.rcParams["animation.ffmpeg_path"] = imageio_ffmpeg.get_ffmpeg_exe()
SRB = SR / BIN
STATE_COL = {na.STATE_CORRECT: C_HIT, na.STATE_INCORRECT: C_MISS,
             na.STATE_ITI: P["single"], na.STATE_REWARD: C_HIT}
STATE_TXT = {na.STATE_CORRECT: "hit trial", na.STATE_INCORRECT: "miss trial",
             na.STATE_ITI: "inter-trial / baseline", na.STATE_REWARD: "reward"}


def bin_mean(x, nb):
    """A 30 Hz trace averaged into the neural bins (NaN-safe)."""
    x = np.asarray(x, float)[:nb * BIN]
    if len(x) < nb * BIN:
        x = np.concatenate([x, np.full(nb * BIN - len(x), np.nan)])
    return np.nanmean(x.reshape(nb, BIN), axis=1)


def trajectory_movie(day_i, out_path):
    """One PCA space, three PC-pair panels, and the cursor on TWO rows -- what the animal
    drove, and the other-axis simulation underneath it. Overlaying them was unreadable."""
    S = session(MOUSE, day_i)
    X, trb = binned(MOUSE, day_i)
    sim = simulate_other_axis(MOUSE, day_i)
    nb = X.shape[1]
    MU_D = X.mean(axis=1)
    _v, vec, vrat = eig_space(np.cov(X))
    ref = float(S.remap)
    w0 = int(max(ref - PRE_MIN * 60 * SR, 0) // BIN)
    w1 = int(min(ref + POST_MIN * 60 * SR, S.nf) // BIN)
    Pr = gaussian_filter1d((X[:, w0:w1].T - MU_D) @ vec[:, :4], SMOOTH_BIN, axis=0,
                           mode="nearest")
    tmin = (np.arange(w0, w1) * BIN - ref) / SR / 60.0
    lab_all = na.label_timepoints(trb, nb, sr=SRB)
    lab = lab_all[w0:w1]
    rewb = np.unique(np.asarray(na.bin_frames(S.rew, BIN), int))
    rews = [int(f - w0) for f in rewb if w0 <= f < w1]
    cur = bin_mean(S.cur_raw, nb)[w0:w1]
    oth = bin_mean(sim_cursor(sim, "other", smooth=False), nb)[w0:w1]
    rl = bin_mean(S.rew_level, nb)[w0:w1]
    fo = bin_mean(np.asarray(sim.fac_other, float), nb)[w0:w1]
    fl = bin_mean(np.asarray(sim.fac, float), nb)[w0:w1]      # which axis was LIVE

    def _axis_col(v):
        """A trial end is coloured by the axis that was driving when the reward landed."""
        if not np.isfinite(v):
            return P["fg"]
        return C_INT if abs(v - F_INT) <= abs(v - F_NEW) else C_NEW

    rew_col = [_axis_col(fl[i]) for i in rews]
    i_rot = int(np.argmin(np.abs(tmin)))
    i_wash = (int(S.blocks[2].start_frame // BIN) - w0) if len(S.blocks) >= 3 else None
    if i_wash is not None and not (0 <= i_wash < len(tmin)):
        i_wash = None
    # the two solution states and the quiescence baseline, as in 5.5
    wN = max(int(round(0.5 * SRB)), 1)

    def ends(k, late_half=False):
        b = S.blocks[k]
        idx = [i for i in range(b.first_trial, b.last_trial + 1) if trb.success[i] == 1]
        if late_half:
            idx = idx[len(idx) // 2:]
        out = [X[:, max(int(trb.end[i]) - wN, int(trb.start[i])):int(trb.end[i])].mean(1)
               for i in idx if int(trb.end[i]) > int(trb.start[i])]
        return np.array(out) if out else np.empty((0, X.shape[0]))

    qf = S.quiet[:nb * BIN].reshape(nb, BIN).mean(1) > 0.5
    q_bins = qf & (lab_all == na.STATE_ITI)
    marks = [((ends(0).mean(0) - MU_D) @ vec[:, :4], C_INT, "X", 150, "$T_{int}$"),
             ((ends(1, True).mean(0) - MU_D) @ vec[:, :4], C_NEW, "X", 150, "$T_{new}$"),
             ((X[:, q_bins].mean(1) - MU_D) @ vec[:, :4], P["fg"], "D", 110,
              "quiescence baseline")]

    step = max(int(round(STEP_SEC * SRB)), 1)
    trail = max(int(round(TRAIL_SEC * SRB)), 2)
    frames = list(range(trail, len(Pr), step))

    fig = plt.figure(figsize=FIGSIZE_V)
    gs = fig.add_gridspec(4, 4, height_ratios=[1.0, 1.0, 1.3, 1.3], hspace=0.55, wspace=0.26)
    axP = [fig.add_subplot(gs[:, 0:2]), fig.add_subplot(gs[2:4, 2]),
           fig.add_subplot(gs[2:4, 3])]
    # the cursor rows live in their own sub-grid, so they can stack tight: the outer hspace
    # has to stay wide for the PC panels' axis labels
    gsC = gs[0:2, 2:4].subgridspec(2, 1, hspace=CUR_HSPACE)
    axR = fig.add_subplot(gsC[0])                         # what the animal drove
    axS = fig.add_subplot(gsC[1], sharex=axR)             # the other-axis simulation
    trails, heads, rmk = [], [], []
    for ax, (a, b) in zip(axP, PC_PAIRS_V):
        pts = Pr[:, [a, b]]
        for q, c_, mk, sz, _lb in marks:
            ax.scatter(q[a], q[b], s=sz, marker=mk, color=c_, edgecolor=P["bg"], lw=0.8,
                       zorder=7)
        # the limits have to hold the markers too, or T_int and the baseline fall off the
        # panel on a short window
        xs = np.concatenate([np.percentile(pts[:, 0], [0.2, 99.8]),
                             [qq[a] for qq, *_ in marks]])
        ys = np.concatenate([np.percentile(pts[:, 1], [0.2, 99.8]),
                             [qq[b] for qq, *_ in marks]])
        px, py = 0.06 * np.ptp(xs), 0.06 * np.ptp(ys)
        ax.set_xlim(xs.min() - px, xs.max() + px)
        ax.set_ylim(ys.min() - py, ys.max() + py)
        ax.set(xlabel=f"PC{a + 1}  ({vrat[a]:.1%})", ylabel=f"PC{b + 1}  ({vrat[b]:.1%})")
        lc = LineCollection([], linewidths=1.8); ax.add_collection(lc); trails.append(lc)
        heads.append(ax.scatter([], [], s=120, zorder=9, edgecolor=P["bg"], linewidth=0.8))
        rmk.append(ax.scatter([], [], s=30, facecolor="none", edgecolor="none", lw=1.2,
                              alpha=ENDMARK_A, zorder=8))
    if LEGEND:
        dw.legend_from(axP[0],
                       [("hit trial", "line", C_HIT, dict(lw=2.2)),
                        ("miss trial", "line", C_MISS, dict(lw=2.2)),
                        ("inter-trial / baseline", "line", P["single"], dict(lw=2.2))] +
                       [(lb, "marker", c_, dict(marker=mk, ms=9)) for _q, c_, mk, _s, lb in marks] +
                       [(f"trial end, last {N_ENDMARK} (coloured by the live axis)",
                         "marker", P["fg"], dict(marker="o", mfc="none"))],
                       loc="upper right", fontsize=11 * FONT_SCALE)

    # ---- row 1: the cursor the animal drove ----
    axR.plot(tmin, cur, color=P["cursor"], lw=0.9, zorder=3)
    axR.plot(tmin, rl, color=P["correct"], lw=1.0, zorder=1)
    if rews:
        axR.scatter(tmin[rews], cur[rews], s=22, facecolor="none", edgecolor=rew_col,
                    lw=1.1, alpha=ENDMARK_A, zorder=5)
    _ttl_kw = dict(ha="left", va="top", color=P["fg"], fontsize=11.5 * FONT_SCALE,
                   zorder=9, bbox=dict(facecolor=P["bg"], edgecolor="none", alpha=0.62,
                                       pad=1.5))
    axR.text(0.012, 0.96, "the cursor the animal drove", transform=axR.transAxes, **_ttl_kw)
    # ---- row 2: the other-axis simulation, coloured by WHICH axis is simulated ----
    for f_, c_ in [(F_INT, C_INT), (F_NEW, C_NEW)]:
        axS.plot(tmin, np.where(np.round(fo) == f_, oth, np.nan), color=c_, lw=0.9, zorder=3)
    # the simulated axis has its OWN reward threshold; in "brain" units that is a different
    # number from the live one, in every other representation the target is the same constant
    rl_sim = (bin_mean(np.asarray(sim.th_other, float), nb)[w0:w1]
              if REP["name"] == "brain" else rl)
    axS.plot(tmin, rl_sim, color=P["correct"], lw=1.0, zorder=1)
    axS.text(0.012, 0.96, "other-axis simulation", transform=axS.transAxes, **_ttl_kw)
    axS.set_xlabel("time from the rotation (min)")
    vlines = []
    rep_axis(axR)
    rep_axis(axS, label=False, ticks=False)   # same scale as row 1; one set of ticks is enough
    axS.set_yticks([]); axS.minorticks_off()
    for ax in (axR, axS):
        ax.axvline(0.0, color=P["fg"], ls="--", lw=1.4, zorder=6)
        if i_wash is not None:
            ax.axvline(tmin[i_wash], color=P["fg"], ls=":", lw=1.4, zorder=6)
        vlines.append(ax.axvline(tmin[0], color=P["accent"], lw=1.6, zorder=7))
    axR.tick_params(labelbottom=False)
    if LEGEND:
        dw.legend_from(axS, [("simulated intuitive", "line", C_INT, dict(lw=1.6)),
                             ("simulated new", "line", C_NEW, dict(lw=1.6))],
                       loc="upper left", ncol=2, fontsize=9.5 * FONT_SCALE)
    half = TS_WIN_MIN / 2.0
    ttl = fig.suptitle("", fontsize=14 * FONT_SCALE, color=P["fg"])

    def update(i):
        s0 = max(i - trail, 0)
        st_ = int(lab[i])
        k_ = [j for j, f in enumerate(rews) if f <= i][-N_ENDMARK:]
        rs, rc = [rews[j] for j in k_], [rew_col[j] for j in k_]
        for lc, hd, rm, (a, b) in zip(trails, heads, rmk, PC_PAIRS_V):
            pts = Pr[:, [a, b]]
            seg = pts[s0:i + 1]
            lc.set_segments(np.stack([seg[:-1], seg[1:]], axis=1) if len(seg) > 1
                            else np.empty((0, 2, 2)))
            lc.set_color([STATE_COL.get(int(x), P["single"]) for x in lab[s0:i]])
            hd.set_offsets(pts[i][None, :]); hd.set_color(STATE_COL.get(st_, P["single"]))
            rm.set_offsets(pts[rs] if rs else np.empty((0, 2)))
            rm.set_edgecolor(rc if rs else "none")
        for vl in vlines:
            vl.set_xdata([tmin[i], tmin[i]])
        axR.set_xlim(tmin[i] - half, tmin[i] + half)      # axS shares this x-axis
        phase = ("INTUITIVE" if i < i_rot else
                 ("NEW AXIS" if (i_wash is None or i < i_wash) else "WASHOUT"))
        ttl.set_text(f"{S.day}  (session {rel_day(MOUSE, day_i)})   "
                     f"t = {tmin[i]:+.2f} min from the rotation   |   block: {phase}   |   "
                     f"{STATE_TXT.get(st_, '')}")
        return (*trails, *heads, *rmk, *vlines, ttl)

    anim = animation.FuncAnimation(fig, update, frames=frames, blit=False)
    anim.save(str(out_path), writer=animation.FFMpegWriter(fps=FPS, bitrate=2600,
                                                           metadata=dict(artist="")),
              dpi=DPI, savefig_kwargs=dict(facecolor=fig.get_facecolor()))
    plt.close(fig)
    return out_path, len(frames)


VID_DIR = st.figdir() / "videos"; VID_DIR.mkdir(exist_ok=True)
if RENDER_VIDEOS:
    for rank in VIDEO_DAYS:
        di = int(REMAP_I[MOUSE][rank])
        p_, n_ = trajectory_movie(di, VID_DIR / f"remap{rank + 1:02d}_session"
                                               f"{rel_day(MOUSE, di)}_{DAYS[MOUSE][di]}.mp4")
        print(f"remapping session #{rank + 1} (session {rel_day(MOUSE, di)}, "
              f"{DAYS[MOUSE][di]}): {p_.name}, {n_} frames, {p_.stat().st_size / 1e6:.1f} MB")
else:
    print("RENDER_VIDEOS = False — set it True to write the movies to "
          f"{VID_DIR.relative_to(st.repo_root())}")
```

```python
ok = []


def check(name, cond, detail=""):
    ok.append(bool(cond))
    print(f"[{'PASS' if cond else 'FAIL'}] {name}" + (f"  — {detail}" if detail else ""))


# ---- the tone axis is the same scale on every animal and day ----
tones, thrs = [], []
for m in MICE:
    for i in HAS_TRIALS[m]:
        ts = ca.get_timeseries_df(R[m], DAYS[m][int(i)])
        dec = ca.get_decoder_df(R[m], DAYS[m][int(i)])
        o = np.asarray(ts["frequencyOverride"], float) > 0.5
        f = np.asarray(dec["freqToOutput"], float)[:len(o)]
        tones.append(np.unique(np.round(f[o[:len(f)]], 1)))
        thrs.append(np.unique(np.round(np.asarray(ts["cursor_threshold_reward"], float), 3)))
check("the reward tone is one constant on every mouse-day",
      all(len(u) == 1 and abs(u[0] - HZ_REWARD) < 0.05 for u in tones),
      f"{len(tones)} sessions, all at {HZ_REWARD:.1f} Hz")
check("thresholds differ across days although the tone does not",
      len({tuple(t[np.isfinite(t)]) for t in thrs}) > 1,
      f"thresholds in use: {sorted({v for t in thrs for v in t[np.isfinite(t)]})}")

# ---- trial table ----
S = session(MOUSE, 1)
ts = ca.get_timeseries_df(R[MOUSE], DAYS[MOUSE][1])
ce = np.asarray(ts["CE_trial"], float) > 0.5
on = np.flatnonzero(np.diff(np.concatenate([[0], ce.astype(int)])) == 1).astype(float)
dmin = np.array([np.min(np.abs(on - s)) for s in S.trials.start])
check("trials.start lands on a CE_trial onset", np.all(dmin <= 1),
      f"max offset {dmin.max():.0f} frame(s) over {len(dmin)} trials")
check("a hit ends at an active reward", len(S.rew) == int(np.sum(S.trials.success == 1)),
      f"{len(S.rew)} rewards vs {int(np.sum(S.trials.success == 1))} hits")

# ---- the 20 s ceiling ----
miss_d = []
for m in MICE:
    for i in LEARN[m]["i"]:
        tr = session(m, int(i)).trials
        d = ((tr.end - tr.start) / SR)[tr.success == 0]
        miss_d.append(d[np.isfinite(d)])
miss_d = np.concatenate(miss_d)
check("a miss is the 20 s timeout, not a measurement",
      np.all((miss_d >= 19.9) & (miss_d <= 21.0)) and np.mean(np.abs(miss_d - 19.967) < 0.05) > 0.99,
      f"{len(miss_d)} misses: {np.mean(np.abs(miss_d - 19.967) < 0.05):.1%} at exactly 19.97 s, "
      f"the rest up to {miss_d.max():.2f} s")

# ---- animals and days ----
check("mouse_0322R has no remapping session", len(REMAP_I["mouse_0322R"]) == 0,
      "every session runs a single factor_to_use")
check("mouse_0403L rotates on every late session", len(REMAP_I["mouse_0403L"]) == 15,
      f"sessions {rel_day('mouse_0403L', REMAP_I['mouse_0403L'][0])}–"
      f"{rel_day('mouse_0403L', REMAP_I['mouse_0403L'][-1])}")
check("day 0 carries no BMI trial on either animal",
      all(int(TAB[m].set_index("i").loc[day_zero_index(m), "n"]) == 0 for m in MICE),
      ", ".join(f"{m}: {DAYS[m][day_zero_index(m)]}" for m in MICE))
check("the intuitive axis is factor 2, not factor 1, on mouse_0403L",
      set(LEARN["mouse_0403L"]["factors"]) == {"2"} and F_INT == 2.0 and F_NEW == 3.0,
      f"mouse_0322R runs on factor(s) {sorted(set(LEARN['mouse_0322R']['factors']))}")
rot = np.array([session(MOUSE, int(i)).remap for i in REMAP_I[MOUSE]]) / SR / 60.0
check("the rotation always falls in the 10th-11th minute",
      np.all((rot > 10.0) & (rot < 11.5)), f"{rot.min():.2f}–{rot.max():.2f} min")
check("every remapping session is intuitive → rotated → washout",
      all([b.factor for b in session(MOUSE, int(i)).blocks] == [F_INT, F_NEW, F_INT]
          for i in REMAP_I[MOUSE]))

# ---- the trace that is plotted is the tone that was played ----
S = session(MOUSE, 1)
g = np.isfinite(S.hz_raw)
check("smoothing leaves the NaN pattern untouched", np.array_equal(g, np.isfinite(S.hz)),
      f"{(~g).mean():.1%} of frames are outside the experiment")
check("smoothing does not shift the trace",
      abs(np.nanmean(S.hz) - np.nanmean(S.hz_raw)) / np.nanmean(S.hz_raw) < 0.01,
      f"mean {np.nanmean(S.hz_raw):.0f} → {np.nanmean(S.hz):.0f} Hz")
check("the reward tone is drawn, not removed",
      np.all(np.abs(S.hz_raw[S.ovr & g] - HZ_REWARD) < 0.05),
      f"{S.ovr[g].mean():.1%} of in-session frames are the clamp, all at {HZ_REWARD:.1f} Hz")

# ---- how the animals actually learn (section 2.1 / 3.1 / 3.2) ----
S1 = session(MOUSE, 1)
sp = (float(np.nanmin(S1.trials.start)), float(np.nanmax(S1.trials.end)))
ed = sp[0] + np.array([0, 1 / 3, 2 / 3, 1.0]) * (sp[1] - sp[0])
d1 = (S1.trials.end - S1.trials.start) / SR
th = [float(np.nanmedian(d1[(S1.trials.start >= ed[k]) & (S1.trials.start < ed[k + 1])
                            & (S1.trials.success == 1)])) for k in range(3)]
check("time to target falls within the very first session", th[2] < 0.75 * th[0],
      f"{th[0]:.2f} → {th[1]:.2f} → {th[2]:.2f} s across thirds of the session")
check("mouse_0403L can already do the task in session 1",
      LEARN["mouse_0403L"]["hit_rate"].iloc[0] > 0.85,
      f"{LEARN['mouse_0403L']['hit_rate'].iloc[0]:.0%} on its first BMI session")
for m in MICE:
    t = LEARN[m]
    sl = stats.linregress(t["rel"], t["hit_rate"])
    slt = stats.linregress(t["rel"], t["t_hit"])
    check(f"{m}: no across-day trend in hit rate or time to target",
          sl.pvalue > 0.05 and slt.pvalue > 0.05,
          f"hit rate {sl.slope*100:+.2f} %/day (p = {sl.pvalue:.2f}), "
          f"time to target {slt.slope:+.3f} s/day (p = {slt.pvalue:.2f})")
t = LEARN["mouse_0322R"]
check("mouse_0322R starts far below where it settles",
      t["hit_rate"].iloc[0] < t["hit_rate"].median() - 0.2,
      f"{t['hit_rate'].iloc[0]:.0%} on session {t['rel'].iloc[0]} vs a median of "
      f"{t['hit_rate'].median():.0%} over the {len(t)} sessions")
t = LEARN[MOUSE].dropna(subset=["weight"])
pw = stats.pearsonr(t["weight"], t["hit_rate"])[1]
ps = stats.linregress(t["rel"], t["hit_rate"]).pvalue
check("across days, performance tracks body weight and not session number",
      pw < 0.05 < ps, f"weight p = {pw:.3f}, session number p = {ps:.2f}")

# ---- the population claims ----
check("the population is not low-dimensional", PR_POOL > 50,
      f"participation ratio {PR_POOL:.0f} over {len(POOL_IDX)} pooled sessions, "
      f"{DIM80} PCs to 80 %, PC1 {VRAT[0]:.1%}")
check("the estimated axis predicts the rig's own M", np.all(AXR2 > 0.8),
      f"R² {AXR2.min():.2f}–{AXR2.max():.2f}")
check("it agrees with the stored weights up to truncation",
      np.all(cos_store > 0.5) and np.all(cos_store <= np.sqrt(0.62) + 0.05),
      f"|cos| {cos_store.min():.2f}–{cos_store.max():.2f}, ceiling {np.sqrt(0.62):.2f}")
check("the decoder axis is not simply PC1",
      np.all(cos_pc1 < 0.6) and np.all(cos_pc1 > 5 / np.sqrt(len(MU))),
      f"|cos| with PC1 {cos_pc1.min():.2f}–{cos_pc1.max():.2f} vs chance "
      f"{1/np.sqrt(len(MU)):.3f}")

# ---- the remapping claims ----
pv = EP.pivot(index="rel", columns="epoch", values=["hit_rate", "t_hit"])
hi, hn = pv["hit_rate"]["intuitive"], pv["hit_rate"]["new"]
ti, tn = pv["t_hit"]["intuitive"], pv["t_hit"]["new"]
check("the FIRST exposure to the rotation is costly", tn.iloc[0] > 3 * ti.iloc[0],
      f"session {pv.index[0]}: {hi.iloc[0]:.0%} → {hn.iloc[0]:.0%}, "
      f"{ti.iloc[0]:.1f} → {tn.iloc[0]:.1f} s ({tn.iloc[0]/ti.iloc[0]:.1f}x)")
check("by the second exposure the cost is gone", hn.iloc[1] > 0.85,
      f"session {pv.index[1]}: {hn.iloc[1]:.0%} over the whole rotated block")
check("so pooled over every remapping session the rotation is undetectable",
      stats.wilcoxon(hi, hn).pvalue > 0.05 and stats.wilcoxon(ti, tn).pvalue > 0.05,
      f"paired p = {stats.wilcoxon(hi, hn).pvalue:.2g} (hit rate), "
      f"{stats.wilcoxon(ti, tn).pvalue:.2g} (time to target)")
# the washout, measured block by block on the first remapping session
Sw = session(MOUSE, int(REMAP_I[MOUSE][0]))
dw_ = (Sw.trials.end - Sw.trials.start) / SR
bst = []
for k in range(len(Sw.blocks)):
    sl = slice(Sw.blocks[k].first_trial, Sw.blocks[k].last_trial + 1)
    h = Sw.trials.success[sl] == 1
    bst.append((float(np.mean(h)), float(np.nanmedian(dw_[sl][h]))))
check("the washout restores control on the first exposure",
      bst[2][0] >= bst[0][0] - 0.01 and bst[2][1] <= bst[0][1] + 0.1,
      f"washout {bst[2][0]:.0%} / {bst[2][1]:.1f} s vs the opening block's "
      f"{bst[0][0]:.0%} / {bst[0][1]:.1f} s")
m_first = block_window(Sw, 1, 0, 15)
m_last = block_window(Sw, 1, 45, 60)
h_first = float(np.mean(Sw.trials.success[m_first] == 1))
h_last = float(np.mean(Sw.trials.success[m_last] == 1))
check("but that first session's recovery is between sessions, not inside it",
      h_last <= h_first + 0.05,
      f"rotated block: {h_first:.0%} over its first 15 min ({int(m_first.sum())} trials), "
      f"{h_last:.0%} over minutes 45–60 ({int(m_last.sum())} trials)")
check("the two axes really are different directions", np.median(ang) > 45,
      f"median {np.median(ang):.0f} deg between the intuitive and rotated estimates, against "
      f"{chance:.0f} deg for two random directions")

# ---- the other-axis simulation ----
Sx = session(MOUSE, int(REMAP_I[MOUSE][0]))
decx = ca.get_decoder_df(R[MOUSE], Sx.day)
Dx = np.asarray(decx["cursor_brain_raw"], float)[:Sx.nf]
CSx = np.asarray(decx["decoder_angles"], float)[:Sx.nf]
gx = np.isfinite(Dx) & np.isfinite(CSx) & (np.abs(CSx) > SIM_CS_MIN)
rt = np.diff(np.r_[0, (~gx & Sx.in_exp).astype(int), 0])
runs = np.flatnonzero(rt == -1) - np.flatnonzero(rt == 1)
check("||M|| is recoverable from the logger almost everywhere",
      np.mean(~gx[Sx.in_exp]) < 0.15 and runs.max() <= 15,
      f"{np.mean(~gx[Sx.in_exp]):.1%} of in-session frames masked, in runs of a median "
      f"{np.median(runs):.0f} and at most {runs.max():.0f} frames — interpolated")
check("the recovered ||M|| is a positive magnitude",
      np.nanmin(Dx[gx] / CSx[gx] ** 3) > 0,
      f"median {np.nanmedian(Dx[gx] / CSx[gx] ** 3):.2f}")
_cov = []
for _i in REMAP_I[MOUSE]:
    _s, _S = simulate_other_axis(MOUSE, int(_i)), session(MOUSE, int(_i))
    _cov.append(float(np.mean(np.isfinite(np.asarray(_s.D_other, float))[_S.in_exp])))
check("the other-axis simulation is defined on every in-session frame",
      min(_cov) > 0.999,
      f"worst session covers {min(_cov):.3%} of its in-experiment frames; the unlabelled "
      f"changeover between blocks now takes the next block's factor")
vals = [simulate_other_axis(MOUSE, int(i)).val for i in REMAP_I[MOUSE]]
vals = [v for v in vals if v]
rM = np.array([v["r_M"] for v in vals]); rHz = np.array([v["r_hz"] for v in vals])
check("the simulation transfers to a block it was not fitted on",
      len(vals) == len(REMAP_I[MOUSE]) and np.median(rM) > 0.85 and rM.min() > 0.75,
      f"intuitive axis fitted on block 0, simulated in the washout: r(M) = "
      f"{rM.min():.2f}–{rM.max():.2f} (median {np.median(rM):.2f}), "
      f"r(tone) = {rHz.min():.2f}–{rHz.max():.2f}")
lo_s = np.array([v["lo_sim"] for v in vals]); lo_r = np.array([v["lo_real"] for v in vals])
check("the simulated cursor is NOT floored at the base tone",
      abs(np.median(lo_s) - np.median(lo_r)) < 0.10,
      f"frames with the cursor below zero: simulated {np.median(lo_s):.1%} vs real "
      f"{np.median(lo_r):.1%} \u2014 a linear Hz axis hides that half of the range, which is "
      f"why CURSOR_REP exists and why \"hz\" is drawn on a log axis")
fs = np.array([v["frac_sim"] for v in vals]); fr = np.array([v["frac_real"] for v in vals])
check("...but the cubic gate inflates its tail, so threshold crossings are an over-estimate",
      np.median(fs) > np.median(fr),
      f"frames at or above the reward tone: simulated {np.median(fs):.2%} vs real "
      f"{np.median(fr):.2%} (medians) — stated on the figure, and the reason 5.1b reads "
      f"the simulated trace for where it sits rather than for rewards")
_th = session(MOUSE, 1).thr
_want = {"hz": HZ_REWARD, "oct": np.log2(HZ_REWARD / 1000.0), "brain": _th, "theta": 1.0}
_got = float(d_to_rep(_th, _th))
check("every cursor representation is available and self-consistent",
      set(REPS) == {"hz", "oct", "brain", "theta"} and
      abs(_got - _want[REP["name"]]) <= 1e-3 * max(abs(_want[REP["name"]]), 1.0),
      f"active: {REP['name']} ({REP['label']}, {REP['scale']} axis); a cursor at theta maps "
      f"to {_got:.4g}, the reward level being {_want[REP['name']]:.4g}")

print(f"\n{sum(ok)}/{len(ok)} checks passed")
```
